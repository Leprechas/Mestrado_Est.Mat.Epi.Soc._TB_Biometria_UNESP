# ANÁLISE INICIAL — TUBERCULOSE SP/RJ 2001–2024

```python
# ============================================================
# ANÁLISE EXPLORATÓRIA — TUBERCULOSE SP/RJ 2001–2024
#
# Estrutura esperada:
# MESTRADO-PROJETO/
# ├── Casos-TB-2001-2024/
# │   ├── Casos-TB-RJ-BRUTO.csv
# │   ├── Casos-TB-RJ-NOVO.csv
# │   ├── Casos-TB-SP-BRUTO.csv
# │   └── Casos-TB-SP-NOVO.csv
# ├── DADOS-BASE-MAPA/
# │   ├── RJ_geojs-33-mun.json
# │   └── SP_geojs-35-mun.json
# └── População/
#     └── PopulaçãoCorrigida_interpolada_CAGR_FINAL.xlsx
# ============================================================

from pathlib import Path
import re
import unicodedata
import warnings
import csv

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURAÇÕES GERAIS
# ============================================================

YEARS = list(range(2001, 2025))

# O notebook deve estar dentro da pasta MESTRADO-PROJETO
BASE_DIR = Path.cwd()

print("Pasta-base detectada:")
print(BASE_DIR)

# ============================================================
# 2. FUNÇÕES AUXILIARES DE CAMINHO E TEXTO
# ============================================================

def norm_text(x):
    """
    Normaliza texto:
    - remove acentos;
    - remove espaços duplicados;
    - converte para maiúsculas.
    """
    if pd.isna(x):
        return ""

    x = str(x).strip()
    x = unicodedata.normalize("NFKD", x)
    x = "".join(ch for ch in x if not unicodedata.combining(ch))
    x = re.sub(r"\s+", " ", x)
    return x.upper().strip()


def normalizar_path_text(x):
    """
    Normalização para busca de arquivos/pastas.
    """
    x = str(x)
    x = unicodedata.normalize("NFKD", x)
    x = "".join(ch for ch in x if not unicodedata.combining(ch))
    x = x.lower()
    return x


def encontrar_pasta(base, nome_esperado):
    """
    Encontra pasta mesmo se houver problema de acento.
    """
    base = Path(base)

    direto = base / nome_esperado
    if direto.exists():
        return direto

    alvo = normalizar_path_text(nome_esperado)

    for p in base.iterdir():
        if p.is_dir() and normalizar_path_text(p.name) == alvo:
            return p

    print("\nPastas encontradas em:", base)
    for p in base.iterdir():
        if p.is_dir():
            print(" -", p.name)

    raise FileNotFoundError(f"Pasta não encontrada: {nome_esperado}")


def procurar_arquivo(pasta, padroes):
    """
    Procura arquivo dentro de uma pasta usando padrões flexíveis.
    """
    pasta = Path(pasta)

    if not pasta.exists():
        raise FileNotFoundError(f"Pasta não encontrada: {pasta}")

    arquivos = [p for p in pasta.rglob("*") if p.is_file()]

    for arq in arquivos:
        nome = normalizar_path_text(arq.name)

        for padrao in padroes:
            padrao_norm = normalizar_path_text(padrao)
            if re.search(padrao_norm, nome):
                return arq

    print(f"\nArquivos encontrados em {pasta}:")
    for arq in arquivos:
        print(" -", arq.name)

    raise FileNotFoundError(
        f"Nenhum arquivo encontrado em {pasta} para os padrões: {padroes}"
    )


def salvar_fig(nome):
    """
    Salva e exibe figura.
    """
    path = OUT_DIR / nome
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    print("Salvo em:", path)

# ============================================================
# 3. PASTAS E ARQUIVOS
# ============================================================

DIR_CASOS = encontrar_pasta(BASE_DIR, "Casos-TB-2001-2024")
DIR_GEO = encontrar_pasta(BASE_DIR, "DADOS-BASE-MAPA")
DIR_POP = encontrar_pasta(BASE_DIR, "População")

# ------------------------------------------------------------
# IMPORTANTE:
# Usar APENAS os arquivos NOVO.
# Os arquivos BRUTO são ignorados.
# ------------------------------------------------------------

ARQ_CASOS_SP = DIR_CASOS / "Casos-TB-SP-NOVO.csv"
ARQ_CASOS_RJ = DIR_CASOS / "Casos-TB-RJ-NOVO.csv"

if not ARQ_CASOS_SP.exists():
    raise FileNotFoundError(f"Arquivo de casos SP não encontrado: {ARQ_CASOS_SP}")

if not ARQ_CASOS_RJ.exists():
    raise FileNotFoundError(f"Arquivo de casos RJ não encontrado: {ARQ_CASOS_RJ}")

# População
ARQ_POPULACAO = DIR_POP / "PopulaçãoCorrigida_interpolada_CAGR_FINAL.xlsx"

if not ARQ_POPULACAO.exists():
    ARQ_POPULACAO = procurar_arquivo(
        DIR_POP,
        [
            r"populacao.*cagr.*final.*\.xlsx$",
            r"populacao.*interpolada.*\.xlsx$",
            r"populacao.*corrigida.*\.xlsx$",
            r".*\.xlsx$"
        ]
    )

# Malhas
ARQ_GEO_RJ = DIR_GEO / "RJ_geojs-33-mun.json"
ARQ_GEO_SP = DIR_GEO / "SP_geojs-35-mun.json"

if not ARQ_GEO_RJ.exists():
    ARQ_GEO_RJ = procurar_arquivo(
        DIR_GEO,
        [
            r"rj.*geo.*\.json$",
            r"33.*mun.*\.json$",
            r"rj.*\.json$"
        ]
    )

if not ARQ_GEO_SP.exists():
    ARQ_GEO_SP = procurar_arquivo(
        DIR_GEO,
        [
            r"sp.*geo.*\.json$",
            r"35.*mun.*\.json$",
            r"sp.*\.json$"
        ]
    )

# Saída
OUT_DIR = BASE_DIR / "saida_exploratoria_TB"
OUT_DIR.mkdir(parents=True, exist_ok=True)

print("\nArquivos que serão usados:")
print("Casos SP:", ARQ_CASOS_SP)
print("Casos RJ:", ARQ_CASOS_RJ)
print("População:", ARQ_POPULACAO)
print("Geo RJ:", ARQ_GEO_RJ)
print("Geo SP:", ARQ_GEO_SP)
print("Saída:", OUT_DIR)

# ============================================================
# 4. FUNÇÕES DE LEITURA E LIMPEZA
# ============================================================

def read_csv_casos(path):
    """
    Lê arquivos NOVO de casos.
    Estrutura esperada:
    Município de notificação;2001;2002;...;2024;Total;descrição
    """
    last_error = None

    for enc in ["latin1", "cp1252", "utf-8-sig", "utf-8"]:
        try:
            df = pd.read_csv(
                path,
                sep=";",
                encoding=enc,
                dtype=str,
                engine="python",
                quoting=csv.QUOTE_NONE,
                keep_default_na=False,
                on_bad_lines="warn"
            )

            df.columns = [
                str(c).replace('"', "").replace("'", "").strip()
                for c in df.columns
            ]

            # remove colunas completamente vazias
            empty_cols = []
            for c in df.columns:
                vals = df[c].astype(str).str.strip()
                if vals.eq("").all():
                    empty_cols.append(c)

            if empty_cols:
                df = df.drop(columns=empty_cols)

            return df

        except Exception as e:
            last_error = e

    raise last_error


def clean_count_series(s):
    """
    Converte colunas de casos para inteiro.
    Trata:
    - '-' como zero;
    - aspas;
    - espaços;
    - ponto de milhar;
    - vírgula decimal.
    """
    s = s.astype(str).str.strip()

    s = (
        s.str.replace('"', "", regex=False)
         .str.replace("'", "", regex=False)
         .str.strip()
    )

    s = s.replace({
        "-": "0",
        "": np.nan,
        "nan": np.nan,
        "NaN": np.nan,
        "None": np.nan
    })

    s = (
        s.str.replace(" ", "", regex=False)
         .str.replace(".", "", regex=False)
         .str.replace(",", ".", regex=False)
    )

    return pd.to_numeric(s, errors="coerce").fillna(0).astype(int)


def clean_pop_series(s):
    """
    Converte população para numérico de forma segura.
    Se já vier como número do Excel, preserva.
    Se vier como texto, limpa separadores.
    """
    if pd.api.types.is_numeric_dtype(s):
        return pd.to_numeric(s, errors="coerce").round()

    txt = s.astype(str).str.strip()

    txt = (
        txt.str.replace('"', "", regex=False)
           .str.replace("'", "", regex=False)
           .str.replace(" ", "", regex=False)
    )

    # Se vier no padrão brasileiro 1.234,56
    mask_brasil = txt.str.contains(",", regex=False)

    txt_brasil = (
        txt[mask_brasil]
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False)
    )

    # Se vier sem vírgula, não remover ponto automaticamente,
    # porque pode ser decimal gerado pelo Excel, tipo 12345.0.
    txt_outros = txt[~mask_brasil]

    out = pd.Series(index=s.index, dtype="float")
    out.loc[mask_brasil] = pd.to_numeric(txt_brasil, errors="coerce")
    out.loc[~mask_brasil] = pd.to_numeric(txt_outros, errors="coerce")

    return out.round()

# ============================================================
# 5. CARREGAMENTO DOS CASOS
# ============================================================

def carregar_casos(path, estado):
    raw = read_csv_casos(path).copy()

    print(f"\nLendo casos {estado}:")
    print(path)
    print("Colunas encontradas:")
    print(raw.columns.tolist())

    # --------------------------------------------------------
    # 5.1 Identifica coluna de município
    # --------------------------------------------------------
    col_mun = None

    # Tentativa 1: coluna com nome contendo MUNICIPIO
    for c in raw.columns:
        if "MUNICIPIO" in norm_text(c):
            col_mun = c
            break

    # Tentativa 2: coluna com valores do tipo "350010 ADAMANTINA"
    if col_mun is None:
        for c in raw.columns:
            serie = (
                raw[c]
                .astype(str)
                .str.replace('"', "", regex=False)
                .str.replace("'", "", regex=False)
                .str.strip()
            )

            n_match = serie.str.match(r"^\d{6}\s+.+", na=False).sum()

            if n_match >= 5:
                col_mun = c
                break

    if col_mun is None:
        print("\nPrévia do arquivo:")
        display(raw.head(10))
        raise ValueError(
            f"Não encontrei coluna de município no arquivo: {path}"
        )

    print("Coluna de município identificada:", col_mun)

    # --------------------------------------------------------
    # 5.2 Limpa coluna de município
    # --------------------------------------------------------
    raw = raw[raw[col_mun].notna()].copy()

    raw[col_mun] = (
        raw[col_mun]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )

    raw = raw[~raw[col_mun].str.upper().isin(["TOTAL", ""])].copy()

    # Extrai código de 6 dígitos e nome
    extracted = raw[col_mun].str.extract(r"^\s*(\d{6})\s+(.+?)\s*$")

    raw["cod_mun6"] = extracted[0]
    raw["municipio"] = extracted[1]

    raw = raw[raw["cod_mun6"].notna()].copy()

    raw["municipio"] = (
        raw["municipio"]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )

    # --------------------------------------------------------
    # 5.3 Identifica colunas de anos
    # --------------------------------------------------------
    year_cols = []

    for c in raw.columns:
        c_str = str(c).replace('"', "").replace("'", "").strip()

        if c_str.isdigit():
            ano = int(c_str)
            if ano in YEARS:
                year_cols.append(c)

    if len(year_cols) == 0:
        print("\nPrévia do arquivo:")
        display(raw.head(10))
        raise ValueError(f"Não encontrei colunas de anos em: {path}")

    print(f"Anos encontrados em {estado}:")
    print([str(c) for c in year_cols])

    # --------------------------------------------------------
    # 5.4 Converte casos
    # --------------------------------------------------------
    for c in year_cols:
        raw[c] = clean_count_series(raw[c])

    # --------------------------------------------------------
    # 5.5 Formato longo município-ano
    # --------------------------------------------------------
    long = raw[["cod_mun6", "municipio"] + year_cols].melt(
        id_vars=["cod_mun6", "municipio"],
        var_name="ano",
        value_name="casos"
    )

    long["ano"] = (
        long["ano"]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .astype(int)
    )

    long["estado"] = estado
    long["municipio"] = long["municipio"].map(norm_text)

    long = long[["estado", "cod_mun6", "municipio", "ano", "casos"]]

    return long


casos_sp = carregar_casos(ARQ_CASOS_SP, "SP")
casos_rj = carregar_casos(ARQ_CASOS_RJ, "RJ")

casos = pd.concat([casos_sp, casos_rj], ignore_index=True)

print("\nCasos carregados:")
display(
    casos.groupby("estado")
    .agg(
        n_linhas=("casos", "size"),
        n_municipios=("cod_mun6", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        casos_total=("casos", "sum"),
        casos_min=("casos", "min"),
        casos_max=("casos", "max"),
        zeros=("casos", lambda x: (x == 0).sum()),
        prop_zeros=("casos", lambda x: (x == 0).mean())
    )
    .reset_index()
)

# ============================================================
# 6. CARREGAMENTO DA POPULAÇÃO
# ============================================================

def encontrar_sheet_estado(xlsx_path, estado):
    """
    Encontra a aba correspondente ao estado.
    Aceita:
    - RJ;
    - Rio de Janeiro;
    - SP;
    - São Paulo.
    """
    xl = pd.ExcelFile(xlsx_path)
    sheets = xl.sheet_names

    estado_norm = norm_text(estado)

    if estado == "RJ":
        termos = ["RJ", "RIO DE JANEIRO"]
    elif estado == "SP":
        termos = ["SP", "SAO PAULO", "SÃO PAULO"]
    else:
        termos = [estado_norm]

    for sh in sheets:
        sh_norm = norm_text(sh)
        if any(norm_text(t) in sh_norm for t in termos):
            return sh

    print("\nAbas disponíveis no arquivo de população:")
    print(sheets)

    raise ValueError(f"Não encontrei aba para {estado} em {xlsx_path}")


def identificar_coluna_codigo(df):
    """
    Identifica coluna de código municipal.
    """
    for c in df.columns:
        cn = norm_text(c)
        if "COD" in cn:
            return c

    # tentativa por conteúdo
    for c in df.columns:
        serie = df[c].astype(str)
        n_match = serie.str.extract(r"(\d{6})")[0].notna().sum()
        if n_match >= 5:
            return c

    return None


def identificar_coluna_municipio(df):
    """
    Identifica coluna de nome do município.
    """
    for c in df.columns:
        cn = norm_text(c)
        if "MUNICIPIO" in cn:
            return c

    for c in df.columns:
        cn = norm_text(c)
        if cn in ["NOME", "NOME_MUNICIPIO", "NM_MUN", "NM_MUNICIPIO"]:
            return c

    return None


def carregar_populacao(path):
    partes = []

    for estado in ["RJ", "SP"]:
        sheet = encontrar_sheet_estado(path, estado)

        print(f"\nLendo população {estado}:")
        print("Aba:", sheet)

        df = pd.read_excel(path, sheet_name=sheet).copy()

        print("Colunas encontradas na população:")
        print(df.columns.tolist())

        col_cod = identificar_coluna_codigo(df)
        col_mun = identificar_coluna_municipio(df)

        if col_cod is None:
            raise ValueError(f"Não encontrei coluna de código municipal na aba {sheet}")

        if col_mun is None:
            raise ValueError(f"Não encontrei coluna de município na aba {sheet}")

        df["cod_mun6"] = df[col_cod].astype(str).str.extract(r"(\d{6})")[0]
        df["municipio"] = df[col_mun].map(norm_text)

        # identifica colunas de anos
        year_cols = []

        for c in df.columns:
            c_str = str(c).strip()

            if c_str.isdigit():
                ano = int(c_str)
                if ano in YEARS:
                    year_cols.append(c)

        if len(year_cols) == 0:
            raise ValueError(f"Não encontrei colunas de anos na aba {sheet}")

        print(f"Anos de população encontrados em {estado}:")
        print([str(c) for c in year_cols])

        long = df[["cod_mun6", "municipio"] + year_cols].melt(
            id_vars=["cod_mun6", "municipio"],
            var_name="ano",
            value_name="populacao"
        )

        long["ano"] = long["ano"].astype(int)
        long["populacao"] = clean_pop_series(long["populacao"])

        long["estado"] = estado

        partes.append(
            long[["estado", "cod_mun6", "municipio", "ano", "populacao"]]
        )

    return pd.concat(partes, ignore_index=True)


populacao = carregar_populacao(ARQ_POPULACAO)

print("\nPopulação carregada:")
display(
    populacao.groupby("estado")
    .agg(
        n_linhas=("populacao", "size"),
        n_municipios=("cod_mun6", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        pop_ausente=("populacao", lambda x: x.isna().sum()),
        pop_min=("populacao", "min"),
        pop_max=("populacao", "max")
    )
    .reset_index()
)

# ============================================================
# 7. AUDITORIA DOS ARQUIVOS DE ENTRADA
# ============================================================

auditoria_casos = (
    casos.groupby("estado")
    .agg(
        linhas=("casos", "size"),
        municipios=("cod_mun6", "nunique"),
        anos=("ano", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        casos_total=("casos", "sum"),
        casos_min=("casos", "min"),
        casos_max=("casos", "max"),
        zeros=("casos", lambda x: (x == 0).sum()),
        prop_zeros=("casos", lambda x: (x == 0).mean())
    )
    .reset_index()
)

auditoria_pop = (
    populacao.groupby("estado")
    .agg(
        linhas=("populacao", "size"),
        municipios=("cod_mun6", "nunique"),
        anos=("ano", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        pop_ausente=("populacao", lambda x: x.isna().sum()),
        pop_min=("populacao", "min"),
        pop_max=("populacao", "max")
    )
    .reset_index()
)

print("\nAuditoria — arquivo de casos:")
display(auditoria_casos)

print("\nAuditoria — arquivo de população:")
display(auditoria_pop)

# ============================================================
# 8. CONSTRUÇÃO DO PAINEL MUNICÍPIO-ANO
# ============================================================

painel = casos.merge(
    populacao,
    on=["estado", "cod_mun6", "ano"],
    how="outer",
    suffixes=("_casos", "_pop"),
    indicator=True
)

painel["municipio"] = painel["municipio_casos"].combine_first(painel["municipio_pop"])

# Se existe população, mas não existe caso no arquivo, assume casos = 0.
painel["casos"] = painel["casos"].fillna(0).astype(int)

painel = painel.sort_values(["estado", "cod_mun6", "ano"]).reset_index(drop=True)

painel["incidencia_100mil"] = np.where(
    painel["populacao"].gt(0),
    painel["casos"] / painel["populacao"] * 100000,
    np.nan
)

painel["log_pop"] = np.where(
    painel["populacao"].gt(0),
    np.log(painel["populacao"]),
    np.nan
)

painel["var_pop_pct"] = (
    painel.groupby(["estado", "cod_mun6"])["populacao"]
    .pct_change() * 100
)

painel = painel[
    [
        "estado",
        "cod_mun6",
        "municipio",
        "ano",
        "casos",
        "populacao",
        "log_pop",
        "incidencia_100mil",
        "var_pop_pct",
        "_merge"
    ]
]

print("\nPainel município-ano:")
display(
    painel.groupby("estado")
    .agg(
        n_linhas=("casos", "size"),
        n_municipios=("cod_mun6", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        casos_total=("casos", "sum"),
        pop_min=("populacao", "min"),
        pop_max=("populacao", "max")
    )
    .reset_index()
)

print("\nDiagnóstico do merge casos × população:")
display(painel["_merge"].value_counts().to_frame("n"))

painel.to_csv(
    OUT_DIR / "base_analitica_TB_SP_RJ_2001_2024.csv",
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 9. AUDITORIA DO PAINEL
# ============================================================

duplicadas = painel[painel.duplicated(["estado", "cod_mun6", "ano"], keep=False)]

auditoria_geral = pd.DataFrame({
    "indicador": [
        "linhas_total",
        "municipios_total",
        "anos_total",
        "duplicatas_estado_municipio_ano",
        "populacao_ausente",
        "populacao_menor_igual_zero",
        "incidencia_ausente",
        "linhas_em_casos_e_populacao",
        "linhas_somente_casos",
        "linhas_somente_populacao"
    ],
    "valor": [
        len(painel),
        painel[["estado", "cod_mun6"]].drop_duplicates().shape[0],
        painel["ano"].nunique(),
        len(duplicadas),
        painel["populacao"].isna().sum(),
        painel["populacao"].le(0).sum(),
        painel["incidencia_100mil"].isna().sum(),
        (painel["_merge"] == "both").sum(),
        (painel["_merge"] == "left_only").sum(),
        (painel["_merge"] == "right_only").sum()
    ]
})

auditoria_estado = (
    painel.groupby("estado")
    .agg(
        n_linhas=("casos", "size"),
        n_municipios=("cod_mun6", "nunique"),
        casos_total=("casos", "sum"),
        populacao_media=("populacao", "mean"),
        incidencia_media_municipio_ano=("incidencia_100mil", "mean"),
        incidencia_mediana_municipio_ano=("incidencia_100mil", "median"),
        incidencia_max=("incidencia_100mil", "max"),
        zeros=("casos", lambda x: (x == 0).sum()),
        prop_zeros=("casos", lambda x: (x == 0).mean())
    )
    .reset_index()
)

print("\nAuditoria geral:")
display(auditoria_geral)

print("\nAuditoria por estado:")
display(auditoria_estado)

# ============================================================
# 10. RESUMO ESTADO-ANO
# ============================================================

resumo_estado_ano = (
    painel.groupby(["estado", "ano"])
    .agg(
        casos_total=("casos", "sum"),
        populacao_total=("populacao", "sum"),
        municipios=("cod_mun6", "nunique"),
        municipios_zero=("casos", lambda x: (x == 0).sum()),
        media_casos_mun=("casos", "mean"),
        var_casos_mun=("casos", "var"),
        media_inc_mun=("incidencia_100mil", "mean"),
        mediana_inc_mun=("incidencia_100mil", "median"),
        p75_inc_mun=("incidencia_100mil", lambda x: x.quantile(0.75)),
        p90_inc_mun=("incidencia_100mil", lambda x: x.quantile(0.90)),
        p95_inc_mun=("incidencia_100mil", lambda x: x.quantile(0.95)),
        p99_inc_mun=("incidencia_100mil", lambda x: x.quantile(0.99)),
        max_inc_mun=("incidencia_100mil", "max")
    )
    .reset_index()
)

resumo_estado_ano["incidencia_100mil_estado"] = (
    resumo_estado_ano["casos_total"] /
    resumo_estado_ano["populacao_total"] * 100000
)

resumo_estado_ano["prop_municipios_zero"] = (
    resumo_estado_ano["municipios_zero"] /
    resumo_estado_ano["municipios"]
)

resumo_estado_ano["indice_sobredispersao_aprox"] = (
    resumo_estado_ano["var_casos_mun"] /
    resumo_estado_ano["media_casos_mun"]
)

print("\nResumo estado-ano:")
display(resumo_estado_ano.head())

resumo_estado_ano.to_csv(
    OUT_DIR / "resumo_estado_ano.csv",
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 11. RESUMO MUNICIPAL DO PERÍODO
# ============================================================

resumo_municipio = (
    painel.groupby(["estado", "cod_mun6", "municipio"])
    .agg(
        casos_total=("casos", "sum"),
        casos_media_anual=("casos", "mean"),
        casos_mediana_anual=("casos", "median"),
        casos_max_anual=("casos", "max"),
        pop_media=("populacao", "mean"),
        pop_min=("populacao", "min"),
        pop_max=("populacao", "max"),
        inc_media=("incidencia_100mil", "mean"),
        inc_mediana=("incidencia_100mil", "median"),
        inc_max=("incidencia_100mil", "max"),
        anos_zero=("casos", lambda x: (x == 0).sum())
    )
    .reset_index()
)

resumo_municipio["prop_anos_zero"] = resumo_municipio["anos_zero"] / len(YEARS)

print("\nTop 20 municípios por casos acumulados:")
display(
    resumo_municipio
    .sort_values("casos_total", ascending=False)
    .head(20)
)

print("\nTop 20 municípios por incidência média:")
display(
    resumo_municipio
    .sort_values("inc_media", ascending=False)
    .head(20)
)

resumo_municipio.to_csv(
    OUT_DIR / "resumo_municipio_periodo.csv",
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# 12. TABELAS DE ALERTA
# ============================================================

POP_PEQUENA = 20_000
INCIDENCIA_PEQUENO_DENOMINADOR = 100
INCIDENCIA_ALTA_FIXA = 300
VAR_POP_ALERTA_PCT = 20

painel_alertas = painel.copy()

painel_alertas["q99_inc_estado"] = (
    painel_alertas.groupby("estado")["incidencia_100mil"]
    .transform(lambda x: x.quantile(0.99))
)

painel_alertas["limiar_alta_incidencia"] = np.maximum(
    INCIDENCIA_ALTA_FIXA,
    painel_alertas["q99_inc_estado"]
)

# Alerta 1: possível problema na população
alerta_populacao = painel_alertas[
    (
        painel_alertas["populacao"].isna()
        | painel_alertas["populacao"].le(0)
        | painel_alertas["var_pop_pct"].abs().gt(VAR_POP_ALERTA_PCT)
    )
].copy()

# Alerta 2: incidência instável por pequeno denominador
alerta_pequeno_denominador = painel_alertas[
    (
        painel_alertas["populacao"].lt(POP_PEQUENA)
        & painel_alertas["casos"].gt(0)
        & painel_alertas["incidencia_100mil"].ge(INCIDENCIA_PEQUENO_DENOMINADOR)
        & ~painel_alertas.index.isin(alerta_populacao.index)
    )
].copy()

# Alerta 3: incidência alta a investigar
alerta_incidencia_alta = painel_alertas[
    (
        painel_alertas["populacao"].ge(POP_PEQUENA)
        & painel_alertas["incidencia_100mil"].ge(painel_alertas["limiar_alta_incidencia"])
        & ~painel_alertas.index.isin(alerta_populacao.index)
    )
].copy()

cols_alerta = [
    "estado",
    "cod_mun6",
    "municipio",
    "ano",
    "casos",
    "populacao",
    "incidencia_100mil",
    "var_pop_pct"
]

alerta_populacao = alerta_populacao[cols_alerta].sort_values(
    ["estado", "municipio", "ano"]
)

alerta_pequeno_denominador = alerta_pequeno_denominador[cols_alerta].sort_values(
    ["incidencia_100mil"], ascending=False
)

alerta_incidencia_alta = alerta_incidencia_alta[cols_alerta].sort_values(
    ["incidencia_100mil"], ascending=False
)

print("\nAlerta 1 — Possível problema na população:")
display(alerta_populacao.head(50))

print("\nAlerta 2 — Incidência instável por pequeno denominador:")
display(alerta_pequeno_denominador.head(50))

print("\nAlerta 3 — Incidência alta a investigar:")
display(alerta_incidencia_alta.head(50))

with pd.ExcelWriter(OUT_DIR / "tabelas_alerta_exploratoria.xlsx") as writer:
    auditoria_geral.to_excel(writer, sheet_name="auditoria_geral", index=False)
    auditoria_estado.to_excel(writer, sheet_name="auditoria_estado", index=False)
    alerta_populacao.to_excel(writer, sheet_name="problema_populacao", index=False)
    alerta_pequeno_denominador.to_excel(writer, sheet_name="pequeno_denominador", index=False)
    alerta_incidencia_alta.to_excel(writer, sheet_name="incidencia_alta", index=False)

# ============================================================
# 13. GRÁFICOS DE SÉRIE TEMPORAL — ESTADO
# ============================================================

plt.figure(figsize=(10, 5))
for estado, sub in resumo_estado_ano.groupby("estado"):
    plt.plot(sub["ano"], sub["casos_total"], marker="o", label=estado)

plt.title("Casos totais de tuberculose por ano — SP e RJ")
plt.xlabel("Ano")
plt.ylabel("Casos")
plt.legend()
plt.grid(alpha=0.3)
salvar_fig("serie_temporal_casos_estado.png")


plt.figure(figsize=(10, 5))
for estado, sub in resumo_estado_ano.groupby("estado"):
    plt.plot(sub["ano"], sub["incidencia_100mil_estado"], marker="o", label=estado)

plt.title("Incidência bruta de tuberculose por 100 mil habitantes — SP e RJ")
plt.xlabel("Ano")
plt.ylabel("Incidência por 100 mil")
plt.legend()
plt.grid(alpha=0.3)
salvar_fig("serie_temporal_incidencia_estado.png")


plt.figure(figsize=(10, 5))
for estado, sub in resumo_estado_ano.groupby("estado"):
    plt.plot(sub["ano"], sub["prop_municipios_zero"], marker="o", label=estado)

plt.title("Proporção de municípios com zero casos por ano")
plt.xlabel("Ano")
plt.ylabel("Proporção de municípios com zero casos")
plt.legend()
plt.grid(alpha=0.3)
salvar_fig("serie_temporal_prop_zeros_estado.png")


plt.figure(figsize=(10, 5))
for estado, sub in resumo_estado_ano.groupby("estado"):
    plt.plot(sub["ano"], sub["indice_sobredispersao_aprox"], marker="o", label=estado)

plt.title("Índice aproximado de sobredispersão por ano")
plt.xlabel("Ano")
plt.ylabel("Variância dos casos municipais / média dos casos municipais")
plt.legend()
plt.grid(alpha=0.3)
salvar_fig("sobredispersao_aproximada_estado_ano.png")

# ============================================================
# 14. DISTRIBUIÇÕES
# ============================================================

for estado in ["SP", "RJ"]:
    sub = painel.query("estado == @estado").copy()

    cap = sub["incidencia_100mil"].quantile(0.99)
    x = sub["incidencia_100mil"].clip(upper=cap)

    plt.figure(figsize=(9, 5))
    plt.hist(x.dropna(), bins=50)
    plt.title(
        f"Distribuição da incidência município-ano — {estado}\n"
        f"valores acima do p99 truncados"
    )
    plt.xlabel("Incidência por 100 mil")
    plt.ylabel("Frequência município-ano")
    plt.grid(alpha=0.3)
    salvar_fig(f"distribuicao_incidencia_municipio_ano_{estado}.png")


for estado in ["SP", "RJ"]:
    sub = painel.query("estado == @estado").copy()

    cap = sub["casos"].quantile(0.99)
    x = sub["casos"].clip(upper=cap)

    plt.figure(figsize=(9, 5))
    plt.hist(x.dropna(), bins=50)
    plt.title(
        f"Distribuição dos casos município-ano — {estado}\n"
        f"valores acima do p99 truncados"
    )
    plt.xlabel("Casos por município-ano")
    plt.ylabel("Frequência município-ano")
    plt.grid(alpha=0.3)
    salvar_fig(f"distribuicao_casos_municipio_ano_{estado}.png")


plt.figure(figsize=(8, 5))
painel.boxplot(
    column="incidencia_100mil",
    by="estado",
    showfliers=False
)
plt.title("Distribuição da incidência município-ano por estado")
plt.suptitle("")
plt.xlabel("Estado")
plt.ylabel("Incidência por 100 mil")
plt.grid(alpha=0.3)
salvar_fig("boxplot_incidencia_por_estado_sem_outliers.png")


plt.figure(figsize=(8, 5))
painel.boxplot(
    column="casos",
    by="estado",
    showfliers=False
)
plt.title("Distribuição dos casos município-ano por estado")
plt.suptitle("")
plt.xlabel("Estado")
plt.ylabel("Casos")
plt.grid(alpha=0.3)
salvar_fig("boxplot_casos_por_estado_sem_outliers.png")

# ============================================================
# 15. SÉRIE TEMPORAL MUNICIPAL
# ============================================================

def plot_serie_municipal(estado=None, cod_mun6=None, municipio=None):
    """
    Plota série temporal de casos e incidência para um município.

    Exemplos:
    plot_serie_municipal(estado="RJ", municipio="RIO DE JANEIRO")
    plot_serie_municipal(estado="SP", municipio="SAO PAULO")
    plot_serie_municipal(cod_mun6="330455")
    """

    df = painel.copy()

    if estado is not None:
        df = df[df["estado"] == estado]

    if cod_mun6 is not None:
        df = df[df["cod_mun6"].astype(str) == str(cod_mun6)]

    if municipio is not None:
        municipio_norm = norm_text(municipio)
        df = df[df["municipio"].str.contains(municipio_norm, na=False)]

    if df.empty:
        print("Nenhum município encontrado.")
        return

    municipios = df[["estado", "cod_mun6", "municipio"]].drop_duplicates()

    if len(municipios) > 1:
        print("Mais de um município encontrado. Refine a busca.")
        display(municipios)
        return

    info = municipios.iloc[0]
    titulo = f"{info['municipio']} ({info['estado']}) — {info['cod_mun6']}"

    df = df.sort_values("ano")

    plt.figure(figsize=(10, 5))
    plt.plot(df["ano"], df["casos"], marker="o")
    plt.title(f"Casos de tuberculose — {titulo}")
    plt.xlabel("Ano")
    plt.ylabel("Casos")
    plt.grid(alpha=0.3)
    salvar_fig(f"serie_casos_{info['estado']}_{info['cod_mun6']}.png")

    plt.figure(figsize=(10, 5))
    plt.plot(df["ano"], df["incidencia_100mil"], marker="o")
    plt.title(f"Incidência por 100 mil — {titulo}")
    plt.xlabel("Ano")
    plt.ylabel("Incidência por 100 mil")
    plt.grid(alpha=0.3)
    salvar_fig(f"serie_incidencia_{info['estado']}_{info['cod_mun6']}.png")

    display(df)


# Exemplos principais
plot_serie_municipal(estado="RJ", municipio="RIO DE JANEIRO")
plot_serie_municipal(estado="SP", municipio="SAO PAULO")

# ============================================================
# 16. SÉRIES DOS MUNICÍPIOS COM MAIS CASOS / INCIDÊNCIA
# ============================================================

def plot_top_municipios_casos(estado, n=5):
    top = (
        resumo_municipio.query("estado == @estado")
        .sort_values("casos_total", ascending=False)
        .head(n)
    )

    plt.figure(figsize=(11, 6))

    for _, row in top.iterrows():
        sub = painel[
            (painel["estado"] == estado)
            & (painel["cod_mun6"] == row["cod_mun6"])
        ].sort_values("ano")

        plt.plot(
            sub["ano"],
            sub["casos"],
            marker="o",
            label=row["municipio"]
        )

    plt.title(f"Casos anuais — top {n} municípios por casos acumulados — {estado}")
    plt.xlabel("Ano")
    plt.ylabel("Casos")
    plt.legend()
    plt.grid(alpha=0.3)
    salvar_fig(f"top_{n}_municipios_casos_{estado}.png")


def plot_top_municipios_incidencia(estado, n=5):
    top = (
        resumo_municipio.query("estado == @estado")
        .sort_values("inc_media", ascending=False)
        .head(n)
    )

    plt.figure(figsize=(11, 6))

    for _, row in top.iterrows():
        sub = painel[
            (painel["estado"] == estado)
            & (painel["cod_mun6"] == row["cod_mun6"])
        ].sort_values("ano")

        plt.plot(
            sub["ano"],
            sub["incidencia_100mil"],
            marker="o",
            label=row["municipio"]
        )

    plt.title(f"Incidência anual — top {n} municípios por incidência média — {estado}")
    plt.xlabel("Ano")
    plt.ylabel("Incidência por 100 mil")
    plt.legend()
    plt.grid(alpha=0.3)
    salvar_fig(f"top_{n}_municipios_incidencia_{estado}.png")


plot_top_municipios_casos("RJ", n=5)
plot_top_municipios_casos("SP", n=5)

plot_top_municipios_incidencia("RJ", n=5)
plot_top_municipios_incidencia("SP", n=5)

# ============================================================
# 17. MAPAS TEMÁTICOS
# ============================================================
# Se necessário, instale:
# !pip install geopandas mapclassify

try:
    import geopandas as gpd

    def carregar_malha(path, estado):
        gdf = gpd.read_file(path).copy()

        if "id" not in gdf.columns:
            raise ValueError(f"A malha {path} não tem coluna 'id'.")

        # GeoJSON usa código de 7 dígitos; base usa os 6 primeiros.
        gdf["cod_mun6"] = gdf["id"].astype(str).str[:6]

        if "name" in gdf.columns:
            gdf["municipio_geo"] = gdf["name"].map(norm_text)
        else:
            gdf["municipio_geo"] = ""

        gdf["estado"] = estado

        return gdf

    geo_rj = carregar_malha(ARQ_GEO_RJ, "RJ")
    geo_sp = carregar_malha(ARQ_GEO_SP, "SP")

    geos = {
        "RJ": geo_rj,
        "SP": geo_sp
    }

    print("\nMalhas carregadas:")
    display(
        pd.DataFrame({
            "estado": ["RJ", "SP"],
            "municipios_malha": [
                geo_rj["cod_mun6"].nunique(),
                geo_sp["cod_mun6"].nunique()
            ]
        })
    )

    def mapa_estado_ano(estado, ano, variavel="incidencia_100mil"):
        gdf = geos[estado].copy()

        dados = painel[
            (painel["estado"] == estado)
            & (painel["ano"] == ano)
        ].copy()

        gdf = gdf.merge(
            dados,
            on=["estado", "cod_mun6"],
            how="left"
        )

        fig, ax = plt.subplots(figsize=(9, 9))

        gdf.plot(
            column=variavel,
            ax=ax,
            legend=True,
            missing_kwds={
                "color": "lightgrey",
                "label": "Sem dado"
            }
        )

        ax.set_title(f"{variavel} — {estado}, {ano}")
        ax.axis("off")

        nome = f"mapa_{variavel}_{estado}_{ano}.png"
        plt.savefig(OUT_DIR / nome, dpi=300, bbox_inches="tight")
        plt.show()

        print("Salvo em:", OUT_DIR / nome)

    def mapa_periodo(estado, variavel="inc_media"):
        gdf = geos[estado].copy()

        dados = resumo_municipio[
            resumo_municipio["estado"] == estado
        ].copy()

        gdf = gdf.merge(
            dados,
            on=["estado", "cod_mun6"],
            how="left"
        )

        fig, ax = plt.subplots(figsize=(9, 9))

        gdf.plot(
            column=variavel,
            ax=ax,
            legend=True,
            missing_kwds={
                "color": "lightgrey",
                "label": "Sem dado"
            }
        )

        ax.set_title(f"{variavel} — {estado}, 2001–2024")
        ax.axis("off")

        nome = f"mapa_{variavel}_{estado}_2001_2024.png"
        plt.savefig(OUT_DIR / nome, dpi=300, bbox_inches="tight")
        plt.show()

        print("Salvo em:", OUT_DIR / nome)

    # Mapas por ano
    mapa_estado_ano("RJ", 2001, "incidencia_100mil")
    mapa_estado_ano("RJ", 2024, "incidencia_100mil")

    mapa_estado_ano("SP", 2001, "incidencia_100mil")
    mapa_estado_ano("SP", 2024, "incidencia_100mil")

    # Mapas do período
    mapa_periodo("RJ", "inc_media")
    mapa_periodo("SP", "inc_media")

except ImportError:
    print("\ngeopandas não está instalado. Pulei os mapas.")
    print("Para instalar, rode:")
    print("!pip install geopandas mapclassify")

except Exception as e:
    print("\nMapas não foram gerados.")
    print("Motivo:", e)

# ============================================================
# 18. MORAN GLOBAL OPCIONAL
# ============================================================
# Se necessário, instale:
# !pip install libpysal esda

try:
    from libpysal.weights import Queen
    from esda.moran import Moran

    def moran_global(estado, ano, variavel="incidencia_100mil", permutations=999):
        gdf = geos[estado].copy()

        dados = painel[
            (painel["estado"] == estado)
            & (painel["ano"] == ano)
        ].copy()

        gdf = gdf.merge(
            dados,
            on=["estado", "cod_mun6"],
            how="left"
        )

        gdf = gdf.dropna(subset=[variavel]).copy()

        w = Queen.from_dataframe(gdf)
        w.transform = "r"

        y = gdf[variavel].values

        moran = Moran(y, w, permutations=permutations)

        resultado = {
            "estado": estado,
            "ano": ano,
            "variavel": variavel,
            "I_Moran": moran.I,
            "p_sim": moran.p_sim,
            "z_sim": moran.z_sim,
            "permutations": permutations
        }

        return resultado

    moran_resultados = []

    for estado in ["RJ", "SP"]:
        for ano in YEARS:
            moran_resultados.append(
                moran_global(estado, ano, "incidencia_100mil")
            )

    moran_df = pd.DataFrame(moran_resultados)

    print("\nMoran Global por estado-ano:")
    display(moran_df.head())

    moran_df.to_csv(
        OUT_DIR / "moran_global_estado_ano.csv",
        index=False,
        encoding="utf-8-sig"
    )

    plt.figure(figsize=(10, 5))
    for estado, sub in moran_df.groupby("estado"):
        plt.plot(sub["ano"], sub["I_Moran"], marker="o", label=estado)

    plt.axhline(0, linestyle="--")
    plt.title("Moran Global da incidência por ano")
    plt.xlabel("Ano")
    plt.ylabel("I de Moran")
    plt.legend()
    plt.grid(alpha=0.3)
    salvar_fig("moran_global_incidencia_estado_ano.png")

except Exception as e:
    print("\nMoran Global não foi executado.")
    print("Motivo:", e)
    print("Se for falta de pacote, rode:")
    print("!pip install libpysal esda")

# ============================================================
# 19. EXPORTAÇÃO FINAL
# ============================================================

with pd.ExcelWriter(OUT_DIR / "analise_exploratoria_TB_SP_RJ_2001_2024.xlsx") as writer:
    auditoria_casos.to_excel(writer, sheet_name="auditoria_casos", index=False)
    auditoria_pop.to_excel(writer, sheet_name="auditoria_populacao", index=False)
    painel.to_excel(writer, sheet_name="base_analitica", index=False)
    auditoria_geral.to_excel(writer, sheet_name="auditoria_geral", index=False)
    auditoria_estado.to_excel(writer, sheet_name="auditoria_estado", index=False)
    resumo_estado_ano.to_excel(writer, sheet_name="resumo_estado_ano", index=False)
    resumo_municipio.to_excel(writer, sheet_name="resumo_municipio", index=False)
    alerta_populacao.to_excel(writer, sheet_name="problema_populacao", index=False)
    alerta_pequeno_denominador.to_excel(writer, sheet_name="pequeno_denominador", index=False)
    alerta_incidencia_alta.to_excel(writer, sheet_name="incidencia_alta", index=False)

print("\n============================================================")
print("ANÁLISE EXPLORATÓRIA FINALIZADA")
print("Arquivos salvos em:")
print(OUT_DIR)
print("============================================================")
```

```python
# ============================================================
# ANÁLISE EXPLORATÓRIA AMPLIADA — TB SP/RJ 2001–2024
# Rodar após o código principal que criou:
# painel, geo_sp, geo_rj, OUT_DIR
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

from IPython.display import display

# ------------------------------------------------------------
# 0. Checagem dos objetos necessários
# ------------------------------------------------------------

for obj in ["painel", "OUT_DIR"]:
    if obj not in globals():
        raise NameError(f"O objeto '{obj}' não existe. Rode primeiro o código principal.")

OUT_DIR = Path(OUT_DIR)
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_DIR_EXT = OUT_DIR / "analise_exploratoria_ampliada"
OUT_DIR_EXT.mkdir(parents=True, exist_ok=True)

YEARS = list(range(2001, 2025))

print("Saídas serão salvas em:")
print(OUT_DIR_EXT)


# ============================================================
# 1. FUNÇÕES AUXILIARES
# ============================================================

def salvar_fig(nome):
    path = OUT_DIR_EXT / nome
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    print("Salvo em:", path)


def longest_run_from_flags(df, flag_col, year_col="ano"):
    """
    Calcula a maior sequência consecutiva de anos com flag=True.
    """
    df = df.sort_values(year_col)
    anos = df.loc[df[flag_col].fillna(False), year_col].tolist()

    if len(anos) == 0:
        return 0

    max_run = 1
    current = 1

    for i in range(1, len(anos)):
        if anos[i] == anos[i - 1] + 1:
            current += 1
            max_run = max(max_run, current)
        else:
            current = 1

    return max_run


def slope_linear(x, y):
    """
    Inclinação linear simples, intercepto e R².
    Ignora NA.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)

    if mask.sum() < 3:
        return pd.Series({
            "slope": np.nan,
            "intercept": np.nan,
            "r2": np.nan
        })

    x = x[mask]
    y = y[mask]

    slope, intercept = np.polyfit(x, y, 1)
    yhat = slope * x + intercept

    ss_res = np.sum((y - yhat) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)

    r2 = np.nan if ss_tot == 0 else 1 - ss_res / ss_tot

    return pd.Series({
        "slope": slope,
        "intercept": intercept,
        "r2": r2
    })


def classificar_porte_pop(pop):
    if pd.isna(pop):
        return np.nan
    elif pop < 5_000:
        return "< 5 mil"
    elif pop < 10_000:
        return "5–10 mil"
    elif pop < 20_000:
        return "10–20 mil"
    elif pop < 50_000:
        return "20–50 mil"
    elif pop < 100_000:
        return "50–100 mil"
    elif pop < 500_000:
        return "100–500 mil"
    else:
        return "> 500 mil"


def adicionar_periodo(ano):
    if 2001 <= ano <= 2005:
        return "2001–2005"
    elif 2006 <= ano <= 2010:
        return "2006–2010"
    elif 2011 <= ano <= 2015:
        return "2011–2015"
    elif 2016 <= ano <= 2019:
        return "2016–2019"
    elif 2020 <= ano <= 2024:
        return "2020–2024"
    else:
        return np.nan


def adicionar_periodo_covid(ano):
    if 2001 <= ano <= 2019:
        return "Pré-COVID 2001–2019"
    elif 2020 <= ano <= 2021:
        return "COVID 2020–2021"
    elif 2022 <= ano <= 2024:
        return "Pós-COVID 2022–2024"
    else:
        return np.nan


# ============================================================
# 2. BASE EXTENDIDA
# ============================================================

painel_ext = painel.copy()

painel_ext["caso_zero"] = painel_ext["casos"].eq(0)
painel_ext["caso_positivo"] = painel_ext["casos"].gt(0)
painel_ext["porte_pop_ano"] = painel_ext["populacao"].apply(classificar_porte_pop)
painel_ext["periodo"] = painel_ext["ano"].apply(adicionar_periodo)
painel_ext["periodo_covid"] = painel_ext["ano"].apply(adicionar_periodo_covid)

# Índice temporal simples para tendências
painel_ext["ano_centralizado"] = painel_ext["ano"] - painel_ext["ano"].min()

# log da incidência para tendências menos sensíveis a extremos
painel_ext["log_inc_mais1"] = np.log1p(painel_ext["incidencia_100mil"])


# ============================================================
# 3. AUDITORIA FINAL DO PAINEL
# ============================================================

esperado_municipios = {
    "SP": 645,
    "RJ": 92
}

auditoria_painel_estado = (
    painel_ext
    .groupby("estado")
    .agg(
        linhas=("casos", "size"),
        municipios=("cod_mun6", "nunique"),
        anos=("ano", "nunique"),
        casos_total=("casos", "sum"),
        pop_ausente=("populacao", lambda x: x.isna().sum()),
        pop_zero_ou_negativa=("populacao", lambda x: x.le(0).sum()),
        incidencia_ausente=("incidencia_100mil", lambda x: x.isna().sum()),
        casos_negativos=("casos", lambda x: x.lt(0).sum())
    )
    .reset_index()
)

auditoria_painel_estado["municipios_esperados"] = auditoria_painel_estado["estado"].map(esperado_municipios)
auditoria_painel_estado["dif_municipios"] = (
    auditoria_painel_estado["municipios"] -
    auditoria_painel_estado["municipios_esperados"]
)

linhas_por_municipio = (
    painel_ext
    .groupby(["estado", "cod_mun6", "municipio"])
    .agg(
        n_linhas=("ano", "size"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        anos_distintos=("ano", "nunique")
    )
    .reset_index()
)

municipios_incompletos = linhas_por_municipio[
    (linhas_por_municipio["n_linhas"] != len(YEARS)) |
    (linhas_por_municipio["anos_distintos"] != len(YEARS))
].copy()

print("\nAuditoria final por estado:")
display(auditoria_painel_estado)

print("\nMunicípios com painel incompleto:")
display(municipios_incompletos)

auditoria_painel_estado.to_csv(
    OUT_DIR_EXT / "auditoria_painel_estado.csv",
    index=False,
    encoding="utf-8-sig"
)

municipios_incompletos.to_csv(
    OUT_DIR_EXT / "municipios_painel_incompleto.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 4. CASOS ACUMULADOS E CONCENTRAÇÃO DOS CASOS
# ============================================================

casos_acumulados_mun = (
    painel_ext
    .groupby(["estado", "cod_mun6", "municipio"], as_index=False)
    .agg(
        casos_acumulados=("casos", "sum"),
        casos_media_anual=("casos", "mean"),
        casos_max_anual=("casos", "max"),
        anos_com_casos=("caso_positivo", "sum"),
        anos_zero=("caso_zero", "sum"),
        pop_media=("populacao", "mean"),
        pop_min=("populacao", "min"),
        pop_max=("populacao", "max"),
        inc_media=("incidencia_100mil", "mean"),
        inc_mediana=("incidencia_100mil", "median"),
        inc_max=("incidencia_100mil", "max")
    )
)

casos_acumulados_mun["porte_pop_media"] = casos_acumulados_mun["pop_media"].apply(classificar_porte_pop)
casos_acumulados_mun["prop_anos_zero"] = casos_acumulados_mun["anos_zero"] / len(YEARS)

# Ranking e participação percentual por estado
casos_acumulados_mun = casos_acumulados_mun.sort_values(
    ["estado", "casos_acumulados"],
    ascending=[True, False]
)

casos_acumulados_mun["rank_casos_estado"] = (
    casos_acumulados_mun
    .groupby("estado")["casos_acumulados"]
    .rank(method="first", ascending=False)
    .astype(int)
)

total_estado = (
    casos_acumulados_mun
    .groupby("estado")["casos_acumulados"]
    .transform("sum")
)

casos_acumulados_mun["perc_casos_estado"] = (
    casos_acumulados_mun["casos_acumulados"] / total_estado * 100
)

casos_acumulados_mun["perc_acumulado_estado"] = (
    casos_acumulados_mun
    .groupby("estado")["perc_casos_estado"]
    .cumsum()
)

print("\nTop 20 municípios por casos acumulados em cada estado:")
display(
    casos_acumulados_mun
    .query("rank_casos_estado <= 20")
    .sort_values(["estado", "rank_casos_estado"])
)

casos_acumulados_mun.to_csv(
    OUT_DIR_EXT / "casos_acumulados_municipio.csv",
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 4.1 Tabela: quantos municípios concentram 50%, 70%, 80%, 90%
# ------------------------------------------------------------

limiares = [50, 70, 80, 90]

linhas_concentracao = []

for estado, sub in casos_acumulados_mun.groupby("estado"):
    sub = sub.sort_values("rank_casos_estado")

    for limiar in limiares:
        n_mun = sub.loc[sub["perc_acumulado_estado"] <= limiar, "cod_mun6"].nunique()

        # garante que inclui o município que ultrapassa o limiar
        if (sub["perc_acumulado_estado"] >= limiar).any():
            n_mun = int(sub[sub["perc_acumulado_estado"] >= limiar]["rank_casos_estado"].min())

        linhas_concentracao.append({
            "estado": estado,
            "limiar_percentual_casos": limiar,
            "n_municipios_necessarios": n_mun,
            "total_municipios_estado": sub["cod_mun6"].nunique(),
            "perc_municipios": n_mun / sub["cod_mun6"].nunique() * 100
        })

concentracao_casos = pd.DataFrame(linhas_concentracao)

print("\nConcentração dos casos:")
display(concentracao_casos)

concentracao_casos.to_csv(
    OUT_DIR_EXT / "concentracao_casos_pareto.csv",
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 4.2 Curva de Pareto por estado
# ------------------------------------------------------------

for estado, sub in casos_acumulados_mun.groupby("estado"):
    sub = sub.sort_values("rank_casos_estado")

    plt.figure(figsize=(10, 5))
    plt.plot(sub["rank_casos_estado"], sub["perc_acumulado_estado"], marker="o", markersize=3)
    plt.axhline(50, linestyle="--")
    plt.axhline(80, linestyle="--")
    plt.axhline(90, linestyle="--")
    plt.title(f"Curva de concentração dos casos acumulados — {estado}, 2001–2024")
    plt.xlabel("Municípios ordenados por casos acumulados")
    plt.ylabel("% acumulado dos casos do estado")
    plt.grid(alpha=0.3)
    salvar_fig(f"pareto_casos_acumulados_{estado}.png")


# ============================================================
# 5. TOP 10 POR ESTADO — SÉRIES ANUAIS E ACUMULADAS
# ============================================================

top10_por_estado = (
    casos_acumulados_mun
    .query("rank_casos_estado <= 10")
    .copy()
    .sort_values(["estado", "rank_casos_estado"])
)

display(top10_por_estado)

top10_por_estado.to_csv(
    OUT_DIR_EXT / "top10_municipios_por_estado.csv",
    index=False,
    encoding="utf-8-sig"
)


def plot_top10_estado(estado, acumulado=True):
    top = top10_por_estado.query("estado == @estado").copy()

    df = painel_ext.merge(
        top[["estado", "cod_mun6", "municipio", "rank_casos_estado"]],
        on=["estado", "cod_mun6", "municipio"],
        how="inner"
    )

    df = df.sort_values(["rank_casos_estado", "ano"])

    if acumulado:
        df["valor_plot"] = df.groupby(["estado", "cod_mun6"])["casos"].cumsum()
        ylabel = "Casos acumulados até o ano"
        titulo_tipo = "casos acumulados"
        nome = f"serie_casos_acumulados_top10_{estado}.png"
    else:
        df["valor_plot"] = df["casos"]
        ylabel = "Casos no ano"
        titulo_tipo = "casos anuais"
        nome = f"serie_casos_anuais_top10_{estado}.png"

    plt.figure(figsize=(13, 7))

    for municipio, sub in df.groupby("municipio"):
        sub = sub.sort_values("ano")
        plt.plot(
            sub["ano"],
            sub["valor_plot"],
            marker="o",
            linewidth=2,
            label=municipio
        )

    plt.title(
        f"Série temporal dos {titulo_tipo} de tuberculose\n"
        f"Top 10 municípios por casos acumulados — {estado}, 2001–2024",
        fontsize=14
    )
    plt.xlabel("Ano")
    plt.ylabel(ylabel)
    plt.grid(alpha=0.3)
    plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
    salvar_fig(nome)

    return df


serie_top10_rj_acum = plot_top10_estado("RJ", acumulado=True)
serie_top10_sp_acum = plot_top10_estado("SP", acumulado=True)

serie_top10_rj_anual = plot_top10_estado("RJ", acumulado=False)
serie_top10_sp_anual = plot_top10_estado("SP", acumulado=False)


# ============================================================
# 6. ANÁLISE POR PORTE POPULACIONAL
# ============================================================

ordem_porte = [
    "< 5 mil",
    "5–10 mil",
    "10–20 mil",
    "20–50 mil",
    "50–100 mil",
    "100–500 mil",
    "> 500 mil"
]

analise_porte_munano = (
    painel_ext
    .groupby(["estado", "porte_pop_ano"])
    .agg(
        linhas_municipio_ano=("casos", "size"),
        municipios=("cod_mun6", "nunique"),
        casos_total=("casos", "sum"),
        casos_media=("casos", "mean"),
        incidencia_media=("incidencia_100mil", "mean"),
        incidencia_mediana=("incidencia_100mil", "median"),
        incidencia_p90=("incidencia_100mil", lambda x: x.quantile(0.90)),
        incidencia_p95=("incidencia_100mil", lambda x: x.quantile(0.95)),
        incidencia_max=("incidencia_100mil", "max"),
        zeros=("caso_zero", "sum"),
        prop_zeros=("caso_zero", "mean")
    )
    .reset_index()
)

analise_porte_munano["porte_pop_ano"] = pd.Categorical(
    analise_porte_munano["porte_pop_ano"],
    categories=ordem_porte,
    ordered=True
)

analise_porte_munano = analise_porte_munano.sort_values(["estado", "porte_pop_ano"])

print("\nAnálise por porte populacional — município-ano:")
display(analise_porte_munano)

analise_porte_munano.to_csv(
    OUT_DIR_EXT / "analise_por_porte_populacional_municipio_ano.csv",
    index=False,
    encoding="utf-8-sig"
)

# Boxplot da incidência por porte e estado
for estado in ["RJ", "SP"]:
    sub = painel_ext.query("estado == @estado").copy()
    sub["porte_pop_ano"] = pd.Categorical(sub["porte_pop_ano"], categories=ordem_porte, ordered=True)

    data = [
        sub.loc[sub["porte_pop_ano"] == porte, "incidencia_100mil"].dropna()
        for porte in ordem_porte
    ]

    plt.figure(figsize=(12, 6))
    plt.boxplot(data, labels=ordem_porte, showfliers=False)
    plt.title(f"Distribuição da incidência por porte populacional — {estado}")
    plt.xlabel("Porte populacional do município-ano")
    plt.ylabel("Incidência por 100 mil habitantes")
    plt.xticks(rotation=45)
    plt.grid(alpha=0.3)
    salvar_fig(f"boxplot_incidencia_por_porte_{estado}.png")


# ============================================================
# 7. PERSISTÊNCIA TEMPORAL DE ALTA INCIDÊNCIA
# ============================================================

# Percentis por estado-ano
limiares_estado_ano = (
    painel_ext
    .groupby(["estado", "ano"])
    .agg(
        p90_inc_estado_ano=("incidencia_100mil", lambda x: x.quantile(0.90)),
        p95_inc_estado_ano=("incidencia_100mil", lambda x: x.quantile(0.95)),
        p99_inc_estado_ano=("incidencia_100mil", lambda x: x.quantile(0.99))
    )
    .reset_index()
)

painel_ext = painel_ext.merge(
    limiares_estado_ano,
    on=["estado", "ano"],
    how="left"
)

painel_ext["acima_p90_inc_estado_ano"] = painel_ext["incidencia_100mil"] >= painel_ext["p90_inc_estado_ano"]
painel_ext["acima_p95_inc_estado_ano"] = painel_ext["incidencia_100mil"] >= painel_ext["p95_inc_estado_ano"]
painel_ext["acima_p99_inc_estado_ano"] = painel_ext["incidencia_100mil"] >= painel_ext["p99_inc_estado_ano"]

persistencia = (
    painel_ext
    .groupby(["estado", "cod_mun6", "municipio"])
    .agg(
        casos_acumulados=("casos", "sum"),
        inc_media=("incidencia_100mil", "mean"),
        inc_max=("incidencia_100mil", "max"),
        anos_com_casos=("caso_positivo", "sum"),
        anos_zero=("caso_zero", "sum"),
        anos_acima_p90=("acima_p90_inc_estado_ano", "sum"),
        anos_acima_p95=("acima_p95_inc_estado_ano", "sum"),
        anos_acima_p99=("acima_p99_inc_estado_ano", "sum"),
        pop_media=("populacao", "mean")
    )
    .reset_index()
)

sequencias = []

for keys, sub in painel_ext.groupby(["estado", "cod_mun6", "municipio"]):
    estado, cod, mun = keys

    sequencias.append({
        "estado": estado,
        "cod_mun6": cod,
        "municipio": mun,
        "maior_seq_com_casos": longest_run_from_flags(sub, "caso_positivo"),
        "maior_seq_acima_p90": longest_run_from_flags(sub, "acima_p90_inc_estado_ano"),
        "maior_seq_acima_p95": longest_run_from_flags(sub, "acima_p95_inc_estado_ano")
    })

sequencias = pd.DataFrame(sequencias)

persistencia = persistencia.merge(
    sequencias,
    on=["estado", "cod_mun6", "municipio"],
    how="left"
)

persistencia["prop_anos_zero"] = persistencia["anos_zero"] / len(YEARS)
persistencia["prop_anos_acima_p90"] = persistencia["anos_acima_p90"] / len(YEARS)
persistencia["prop_anos_acima_p95"] = persistencia["anos_acima_p95"] / len(YEARS)

print("\nMunicípios com maior persistência acima do p95 estadual anual:")
display(
    persistencia
    .sort_values(["estado", "anos_acima_p95", "casos_acumulados"], ascending=[True, False, False])
    .groupby("estado")
    .head(20)
)

persistencia.to_csv(
    OUT_DIR_EXT / "persistencia_temporal_alta_incidencia.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 8. TENDÊNCIA TEMPORAL MUNICIPAL
# ============================================================

tend_casos = (
    painel_ext
    .groupby(["estado", "cod_mun6", "municipio"])
    .apply(lambda d: slope_linear(d["ano"], d["casos"]))
    .reset_index()
    .rename(columns={
        "slope": "slope_casos_ano",
        "intercept": "intercept_casos",
        "r2": "r2_casos"
    })
)

tend_inc = (
    painel_ext
    .groupby(["estado", "cod_mun6", "municipio"])
    .apply(lambda d: slope_linear(d["ano"], d["incidencia_100mil"]))
    .reset_index()
    .rename(columns={
        "slope": "slope_incidencia_ano",
        "intercept": "intercept_incidencia",
        "r2": "r2_incidencia"
    })
)

tend_log_inc = (
    painel_ext
    .groupby(["estado", "cod_mun6", "municipio"])
    .apply(lambda d: slope_linear(d["ano"], d["log_inc_mais1"]))
    .reset_index()
    .rename(columns={
        "slope": "slope_log_inc_ano",
        "intercept": "intercept_log_inc",
        "r2": "r2_log_inc"
    })
)

tendencias_municipais = (
    tend_casos
    .merge(tend_inc, on=["estado", "cod_mun6", "municipio"], how="left")
    .merge(tend_log_inc, on=["estado", "cod_mun6", "municipio"], how="left")
    .merge(
        casos_acumulados_mun[[
            "estado", "cod_mun6", "municipio",
            "casos_acumulados", "pop_media", "inc_media", "inc_max"
        ]],
        on=["estado", "cod_mun6", "municipio"],
        how="left"
    )
)

print("\nTop municípios com maior crescimento anual absoluto de casos:")
display(
    tendencias_municipais
    .sort_values(["estado", "slope_casos_ano"], ascending=[True, False])
    .groupby("estado")
    .head(15)
)

print("\nTop municípios com maior queda anual absoluta de casos:")
display(
    tendencias_municipais
    .sort_values(["estado", "slope_casos_ano"], ascending=[True, True])
    .groupby("estado")
    .head(15)
)

tendencias_municipais.to_csv(
    OUT_DIR_EXT / "tendencias_temporais_municipais.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 9. COMPARAÇÃO PRÉ-COVID, COVID E PÓS-COVID
# ============================================================

covid_estado = (
    painel_ext
    .dropna(subset=["periodo_covid"])
    .groupby(["estado", "periodo_covid"])
    .agg(
        casos_total=("casos", "sum"),
        pop_pessoa_ano=("populacao", "sum"),
        municipios=("cod_mun6", "nunique"),
        media_anual_casos_municipio=("casos", "mean"),
        media_inc_mun=("incidencia_100mil", "mean"),
        mediana_inc_mun=("incidencia_100mil", "median"),
        prop_zeros=("caso_zero", "mean")
    )
    .reset_index()
)

covid_estado["incidencia_periodo_100mil_ano"] = (
    covid_estado["casos_total"] / covid_estado["pop_pessoa_ano"] * 100000
)

print("\nComparação por período COVID — estado:")
display(covid_estado)

covid_municipio = (
    painel_ext
    .dropna(subset=["periodo_covid"])
    .groupby(["estado", "cod_mun6", "municipio", "periodo_covid"])
    .agg(
        casos_total=("casos", "sum"),
        pop_pessoa_ano=("populacao", "sum"),
        anos=("ano", "nunique"),
        media_anual_casos=("casos", "mean"),
        media_inc=("incidencia_100mil", "mean"),
        anos_zero=("caso_zero", "sum")
    )
    .reset_index()
)

covid_municipio["incidencia_periodo_100mil_ano"] = (
    covid_municipio["casos_total"] / covid_municipio["pop_pessoa_ano"] * 100000
)

# Tabela larga para comparar períodos por município
covid_wide = covid_municipio.pivot_table(
    index=["estado", "cod_mun6", "municipio"],
    columns="periodo_covid",
    values=["casos_total", "incidencia_periodo_100mil_ano"],
    aggfunc="first"
)

covid_wide.columns = [
    f"{var}_{periodo}".replace(" ", "_").replace("–", "_")
    for var, periodo in covid_wide.columns
]

covid_wide = covid_wide.reset_index()

# Variação pós vs pré, quando possível
col_pre = "incidencia_periodo_100mil_ano_Pré-COVID_2001_2019"
col_pos = "incidencia_periodo_100mil_ano_Pós-COVID_2022_2024"

if col_pre in covid_wide.columns and col_pos in covid_wide.columns:
    covid_wide["dif_inc_pos_menos_pre"] = covid_wide[col_pos] - covid_wide[col_pre]
    covid_wide["razao_inc_pos_pre"] = covid_wide[col_pos] / covid_wide[col_pre].replace(0, np.nan)

print("\nMunicípios com maior aumento pós-COVID vs pré-COVID:")
if "dif_inc_pos_menos_pre" in covid_wide.columns:
    display(
        covid_wide
        .sort_values(["estado", "dif_inc_pos_menos_pre"], ascending=[True, False])
        .groupby("estado")
        .head(20)
    )

covid_estado.to_csv(
    OUT_DIR_EXT / "comparacao_periodos_covid_estado.csv",
    index=False,
    encoding="utf-8-sig"
)

covid_municipio.to_csv(
    OUT_DIR_EXT / "comparacao_periodos_covid_municipio.csv",
    index=False,
    encoding="utf-8-sig"
)

covid_wide.to_csv(
    OUT_DIR_EXT / "comparacao_periodos_covid_municipio_wide.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 10. ZEROS E SOBREDISPERSÃO
# ============================================================

zeros_sobredispersao_estado_ano = (
    painel_ext
    .groupby(["estado", "ano"])
    .agg(
        municipios=("cod_mun6", "nunique"),
        casos_total=("casos", "sum"),
        media_casos_mun=("casos", "mean"),
        var_casos_mun=("casos", "var"),
        municipios_zero=("caso_zero", "sum"),
        prop_municipios_zero=("caso_zero", "mean"),
        media_pop=("populacao", "mean")
    )
    .reset_index()
)

zeros_sobredispersao_estado_ano["razao_var_media"] = (
    zeros_sobredispersao_estado_ano["var_casos_mun"] /
    zeros_sobredispersao_estado_ano["media_casos_mun"].replace(0, np.nan)
)

print("\nZeros e sobredispersão por estado-ano:")
display(zeros_sobredispersao_estado_ano.head())

zeros_sobredispersao_estado_ano.to_csv(
    OUT_DIR_EXT / "zeros_sobredispersao_estado_ano.csv",
    index=False,
    encoding="utf-8-sig"
)

plt.figure(figsize=(10, 5))
for estado, sub in zeros_sobredispersao_estado_ano.groupby("estado"):
    plt.plot(sub["ano"], sub["prop_municipios_zero"], marker="o", label=estado)

plt.title("Proporção de municípios com zero casos por ano")
plt.xlabel("Ano")
plt.ylabel("Proporção de municípios com zero casos")
plt.legend()
plt.grid(alpha=0.3)
salvar_fig("prop_municipios_zero_estado_ano.png")


plt.figure(figsize=(10, 5))
for estado, sub in zeros_sobredispersao_estado_ano.groupby("estado"):
    plt.plot(sub["ano"], sub["razao_var_media"], marker="o", label=estado)

plt.title("Razão variância/média dos casos municipais por ano")
plt.xlabel("Ano")
plt.ylabel("Variância / média")
plt.legend()
plt.grid(alpha=0.3)
salvar_fig("razao_variancia_media_estado_ano.png")


# ============================================================
# 11. INCIDÊNCIA SUAVIZADA SIMPLES E POR PERÍODOS
# ============================================================

# Média móvel de 3 anos da incidência municipal
painel_ext = painel_ext.sort_values(["estado", "cod_mun6", "ano"])

painel_ext["incidencia_ma3"] = (
    painel_ext
    .groupby(["estado", "cod_mun6"])["incidencia_100mil"]
    .transform(lambda x: x.rolling(window=3, min_periods=1, center=True).mean())
)

# Agregação por períodos
periodos_municipio = (
    painel_ext
    .dropna(subset=["periodo"])
    .groupby(["estado", "cod_mun6", "municipio", "periodo"])
    .agg(
        casos_total=("casos", "sum"),
        pop_pessoa_ano=("populacao", "sum"),
        pop_media=("populacao", "mean"),
        anos=("ano", "nunique"),
        inc_media_anual=("incidencia_100mil", "mean"),
        inc_mediana_anual=("incidencia_100mil", "median"),
        anos_zero=("caso_zero", "sum")
    )
    .reset_index()
)

periodos_municipio["incidencia_periodo_100mil_ano"] = (
    periodos_municipio["casos_total"] /
    periodos_municipio["pop_pessoa_ano"] * 100000
)

periodos_estado = (
    painel_ext
    .dropna(subset=["periodo"])
    .groupby(["estado", "periodo"])
    .agg(
        casos_total=("casos", "sum"),
        pop_pessoa_ano=("populacao", "sum"),
        municipios=("cod_mun6", "nunique")
    )
    .reset_index()
)

periodos_estado["incidencia_periodo_100mil_ano"] = (
    periodos_estado["casos_total"] /
    periodos_estado["pop_pessoa_ano"] * 100000
)

print("\nIncidência por período — estado:")
display(periodos_estado)

periodos_municipio.to_csv(
    OUT_DIR_EXT / "incidencia_por_periodo_municipio.csv",
    index=False,
    encoding="utf-8-sig"
)

periodos_estado.to_csv(
    OUT_DIR_EXT / "incidencia_por_periodo_estado.csv",
    index=False,
    encoding="utf-8-sig"
)

plt.figure(figsize=(10, 5))
for estado, sub in periodos_estado.groupby("estado"):
    plt.plot(sub["periodo"], sub["incidencia_periodo_100mil_ano"], marker="o", label=estado)

plt.title("Incidência por período — SP e RJ")
plt.xlabel("Período")
plt.ylabel("Incidência por 100 mil pessoa-ano")
plt.legend()
plt.grid(alpha=0.3)
plt.xticks(rotation=45)
salvar_fig("incidencia_por_periodo_estado.png")


# ============================================================
# 12. SIR EXPLORATÓRIO
# ============================================================

# Taxa estadual anual
taxa_estado_ano = (
    painel_ext
    .groupby(["estado", "ano"])
    .agg(
        casos_estado_ano=("casos", "sum"),
        pop_estado_ano=("populacao", "sum")
    )
    .reset_index()
)

taxa_estado_ano["taxa_estado_ano"] = (
    taxa_estado_ano["casos_estado_ano"] /
    taxa_estado_ano["pop_estado_ano"]
)

painel_ext = painel_ext.merge(
    taxa_estado_ano[["estado", "ano", "taxa_estado_ano"]],
    on=["estado", "ano"],
    how="left"
)

painel_ext["casos_esperados_estado_ano"] = (
    painel_ext["populacao"] * painel_ext["taxa_estado_ano"]
)

painel_ext["sir_bruto"] = (
    painel_ext["casos"] /
    painel_ext["casos_esperados_estado_ano"].replace(0, np.nan)
)

sir_municipio = (
    painel_ext
    .groupby(["estado", "cod_mun6", "municipio"])
    .agg(
        casos_observados=("casos", "sum"),
        casos_esperados=("casos_esperados_estado_ano", "sum"),
        pop_media=("populacao", "mean"),
        inc_media=("incidencia_100mil", "mean")
    )
    .reset_index()
)

sir_municipio["sir_periodo"] = (
    sir_municipio["casos_observados"] /
    sir_municipio["casos_esperados"].replace(0, np.nan)
)

print("\nTop municípios por SIR exploratório do período:")
display(
    sir_municipio
    .sort_values(["estado", "sir_periodo"], ascending=[True, False])
    .groupby("estado")
    .head(20)
)

sir_municipio.to_csv(
    OUT_DIR_EXT / "sir_exploratorio_municipio.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 13. MAPAS — CASOS ACUMULADOS, PERÍODOS, PERSISTÊNCIA E SIR
# ============================================================

try:
    import geopandas as gpd

    if "geo_sp" not in globals() or "geo_rj" not in globals():
        raise NameError("geo_sp e/ou geo_rj não existem. Rode a etapa de mapas do código principal.")

    geo_sp_map = geo_sp.copy()
    geo_rj_map = geo_rj.copy()

    if geo_sp_map.crs is None:
        geo_sp_map = geo_sp_map.set_crs("EPSG:4326")

    if geo_rj_map.crs is None:
        geo_rj_map = geo_rj_map.set_crs("EPSG:4326")

    geo_rj_map = geo_rj_map.to_crs(geo_sp_map.crs)

    geo_dois_estados = pd.concat([geo_sp_map, geo_rj_map], ignore_index=True)

    def plot_mapa_municipal(df_dados, variavel, titulo, nome_arquivo, log1p=False):
        dados = df_dados.copy()

        coluna_plot = variavel

        if log1p:
            coluna_plot = f"{variavel}_log1p"
            dados[coluna_plot] = np.log1p(dados[variavel])

        gdf = geo_dois_estados.merge(
            dados,
            on=["estado", "cod_mun6"],
            how="left"
        )

        fig, ax = plt.subplots(figsize=(12, 10))

        gdf.plot(
            column=coluna_plot,
            ax=ax,
            legend=True,
            linewidth=0.15,
            edgecolor="black",
            missing_kwds={
                "color": "lightgrey",
                "label": "Sem dado"
            }
        )

        ax.set_title(titulo, fontsize=14)
        ax.axis("off")

        plt.savefig(OUT_DIR_EXT / nome_arquivo, dpi=300, bbox_inches="tight")
        plt.show()

        print("Mapa salvo em:", OUT_DIR_EXT / nome_arquivo)

    # Mapa de casos acumulados linear e log
    plot_mapa_municipal(
        casos_acumulados_mun,
        "casos_acumulados",
        "Casos acumulados de tuberculose por município\nSP e RJ, 2001–2024",
        "mapa_casos_acumulados_SP_RJ_linear.png",
        log1p=False
    )

    plot_mapa_municipal(
        casos_acumulados_mun,
        "casos_acumulados",
        "Casos acumulados de tuberculose por município — log(1 + casos)\nSP e RJ, 2001–2024",
        "mapa_casos_acumulados_SP_RJ_log1p.png",
        log1p=True
    )

    # Mapa de persistência acima do p95
    plot_mapa_municipal(
        persistencia,
        "anos_acima_p95",
        "Persistência de alta incidência: anos acima do p95 estadual anual\nSP e RJ, 2001–2024",
        "mapa_anos_acima_p95_SP_RJ.png",
        log1p=False
    )

    # Mapa de SIR
    plot_mapa_municipal(
        sir_municipio,
        "sir_periodo",
        "SIR exploratório municipal\nSP e RJ, 2001–2024",
        "mapa_sir_exploratorio_SP_RJ.png",
        log1p=False
    )

    # Mapas por período e estado
    for periodo in periodos_municipio["periodo"].dropna().unique():
        dados_periodo = periodos_municipio.query("periodo == @periodo").copy()

        plot_mapa_municipal(
            dados_periodo,
            "incidencia_periodo_100mil_ano",
            f"Incidência por 100 mil pessoa-ano — {periodo}\nSP e RJ",
            f"mapa_incidencia_periodo_{periodo.replace('–', '_')}.png",
            log1p=False
        )

except Exception as e:
    print("\nMapas ampliados não foram gerados.")
    print("Motivo:", e)


# ============================================================
# 14. MORAN GLOBAL — EXPLORATÓRIO
# ============================================================

try:
    from libpysal.weights import Queen
    from esda.moran import Moran

    if "geo_sp" not in globals() or "geo_rj" not in globals():
        raise NameError("geo_sp e/ou geo_rj não existem.")

    geos = {
        "RJ": geo_rj.copy(),
        "SP": geo_sp.copy()
    }

    def moran_global_df(estado, dados, variavel):
        gdf = geos[estado].copy()

        if gdf.crs is None:
            gdf = gdf.set_crs("EPSG:4326")

        dados_estado = dados.query("estado == @estado").copy()

        gdf = gdf.merge(
            dados_estado,
            on=["estado", "cod_mun6"],
            how="left"
        )

        gdf = gdf.dropna(subset=[variavel]).copy()

        w = Queen.from_dataframe(gdf)
        w.transform = "r"

        y = gdf[variavel].values

        moran = Moran(y, w, permutations=999)

        return {
            "estado": estado,
            "variavel": variavel,
            "I_Moran": moran.I,
            "p_sim": moran.p_sim,
            "z_sim": moran.z_sim,
            "n": len(gdf)
        }

    moran_periodo_total = []

    for estado in ["RJ", "SP"]:
        moran_periodo_total.append(
            moran_global_df(
                estado,
                casos_acumulados_mun,
                "casos_acumulados"
            )
        )

        moran_periodo_total.append(
            moran_global_df(
                estado,
                sir_municipio,
                "sir_periodo"
            )
        )

        moran_periodo_total.append(
            moran_global_df(
                estado,
                persistencia,
                "anos_acima_p95"
            )
        )

    moran_periodo_total = pd.DataFrame(moran_periodo_total)

    print("\nMoran Global — variáveis agregadas do período:")
    display(moran_periodo_total)

    moran_periodo_total.to_csv(
        OUT_DIR_EXT / "moran_global_variaveis_agregadas.csv",
        index=False,
        encoding="utf-8-sig"
    )

except Exception as e:
    print("\nMoran Global agregado não foi executado.")
    print("Motivo:", e)
    print("Se for falta de pacote, rode:")
    print("!pip install libpysal esda")


# ============================================================
# 15. LISA EXPLORATÓRIO — SIR DO PERÍODO
# ============================================================

try:
    from libpysal.weights import Queen
    from esda.moran import Moran_Local

    def lisa_estado(estado, dados, variavel, nome_saida):
        gdf = geos[estado].copy()

        if gdf.crs is None:
            gdf = gdf.set_crs("EPSG:4326")

        dados_estado = dados.query("estado == @estado").copy()

        gdf = gdf.merge(
            dados_estado,
            on=["estado", "cod_mun6"],
            how="left"
        )

        gdf = gdf.dropna(subset=[variavel]).copy()

        w = Queen.from_dataframe(gdf)
        w.transform = "r"

        y = gdf[variavel].values

        lisa = Moran_Local(y, w, permutations=999)

        gdf["lisa_I"] = lisa.Is
        gdf["lisa_p"] = lisa.p_sim
        gdf["lisa_q"] = lisa.q

        # Classificação: q usa quadrantes
        # 1 = High-High; 2 = Low-High; 3 = Low-Low; 4 = High-Low
        def classificar_lisa(row):
            if row["lisa_p"] > 0.05:
                return "Não significativo"
            elif row["lisa_q"] == 1:
                return "Alto-Alto"
            elif row["lisa_q"] == 2:
                return "Baixo-Alto"
            elif row["lisa_q"] == 3:
                return "Baixo-Baixo"
            elif row["lisa_q"] == 4:
                return "Alto-Baixo"
            else:
                return "Não classificado"

        gdf["classe_lisa"] = gdf.apply(classificar_lisa, axis=1)

        tabela = gdf.drop(columns="geometry").copy()

        tabela.to_csv(
            OUT_DIR_EXT / f"{nome_saida}_{estado}.csv",
            index=False,
            encoding="utf-8-sig"
        )

        # Mapa simples categórico
        fig, ax = plt.subplots(figsize=(9, 9))

        gdf.plot(
            column="classe_lisa",
            ax=ax,
            legend=True,
            linewidth=0.15,
            edgecolor="black"
        )

        ax.set_title(f"LISA exploratório — {variavel} — {estado}")
        ax.axis("off")

        plt.savefig(
            OUT_DIR_EXT / f"mapa_{nome_saida}_{estado}.png",
            dpi=300,
            bbox_inches="tight"
        )

        plt.show()

        print("LISA salvo:", OUT_DIR_EXT / f"{nome_saida}_{estado}.csv")

        return tabela

    lisa_sir_rj = lisa_estado("RJ", sir_municipio, "sir_periodo", "lisa_sir_periodo")
    lisa_sir_sp = lisa_estado("SP", sir_municipio, "sir_periodo", "lisa_sir_periodo")

except Exception as e:
    print("\nLISA não foi executado.")
    print("Motivo:", e)
    print("Se for falta de pacote, rode:")
    print("!pip install libpysal esda")


# ============================================================
# 16. EXPORTAÇÃO GERAL EM EXCEL
# ============================================================

excel_saida = OUT_DIR_EXT / "analise_exploratoria_ampliada_TB_SP_RJ.xlsx"

with pd.ExcelWriter(excel_saida) as writer:
    auditoria_painel_estado.to_excel(writer, sheet_name="auditoria_estado", index=False)
    municipios_incompletos.to_excel(writer, sheet_name="painel_incompleto", index=False)
    casos_acumulados_mun.to_excel(writer, sheet_name="casos_acumulados", index=False)
    concentracao_casos.to_excel(writer, sheet_name="concentracao", index=False)
    top10_por_estado.to_excel(writer, sheet_name="top10_estado", index=False)
    analise_porte_munano.to_excel(writer, sheet_name="porte_pop", index=False)
    persistencia.to_excel(writer, sheet_name="persistencia", index=False)
    tendencias_municipais.to_excel(writer, sheet_name="tendencias", index=False)
    covid_estado.to_excel(writer, sheet_name="covid_estado", index=False)
    covid_municipio.to_excel(writer, sheet_name="covid_municipio", index=False)
    zeros_sobredispersao_estado_ano.to_excel(writer, sheet_name="zeros_sobredisp", index=False)
    periodos_estado.to_excel(writer, sheet_name="periodos_estado", index=False)
    periodos_municipio.to_excel(writer, sheet_name="periodos_municipio", index=False)
    sir_municipio.to_excel(writer, sheet_name="sir_municipio", index=False)

print("\n============================================================")
print("ANÁLISE EXPLORATÓRIA AMPLIADA FINALIZADA")
print("Arquivo Excel principal:")
print(excel_saida)
print("Pasta de saídas:")
print(OUT_DIR_EXT)
print("============================================================")
```
