# ANÁLISE DESCRITIVA E COMPARAÇÃO EPIDEMIOLÓGICA

```python
# ============================================================
# SEÇÃO — ANÁLISE DESCRITIVA E COMPARAÇÃO EPIDEMIOLÓGICA
# COM RECONSTRUÇÃO COMPLETA DO PAINEL
#
# Tuberculose — São Paulo e Rio de Janeiro, 2001–2024
# ============================================================

from pathlib import Path
import re
import csv
import unicodedata
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

warnings.filterwarnings("ignore")

# ============================================================
# 0. CONFIGURAÇÕES GERAIS
# ============================================================

YEARS = list(range(2001, 2025))

COL_ESTADO = "estado"
COL_COD = "cod_mun6"
COL_MUN = "municipio"
COL_ANO = "ano"
COL_CASOS = "casos"
COL_POP = "populacao"
COL_INC = "incidencia_100mil"


# ============================================================
# 1. FUNÇÕES AUXILIARES GERAIS
# ============================================================

def norm_text(x):
    """
    Normaliza texto:
    - remove acentos;
    - remove espaços duplicados;
    - converte para maiúsculas.
    """
    if pd.isna(x):
        return ""

    x = str(x).strip()
    x = unicodedata.normalize("NFKD", x)
    x = "".join(ch for ch in x if not unicodedata.combining(ch))
    x = re.sub(r"\s+", " ", x)
    return x.upper().strip()


def normalizar_path_text(x):
    """
    Normaliza texto para busca de arquivos e pastas.
    """
    x = str(x)
    x = unicodedata.normalize("NFKD", x)
    x = "".join(ch for ch in x if not unicodedata.combining(ch))
    x = x.lower()
    return x


def encontrar_base_dir():
    """
    Procura a pasta-base subindo a partir do diretório atual.
    A pasta-base deve conter:
    - Casos-TB-2001-2024
    - DADOS-BASE-MAPA
    - População
    """
    atual = Path.cwd()

    candidatos = [atual] + list(atual.parents)

    for base in candidatos:
        nomes = [normalizar_path_text(p.name) for p in base.iterdir() if p.is_dir()]

        tem_casos = normalizar_path_text("Casos-TB-2001-2024") in nomes
        tem_mapa = normalizar_path_text("DADOS-BASE-MAPA") in nomes
        tem_pop = normalizar_path_text("População") in nomes

        if tem_casos and tem_mapa and tem_pop:
            return base

    print("Diretório atual:", atual)
    print("Pastas encontradas no diretório atual:")
    for p in atual.iterdir():
        if p.is_dir():
            print(" -", p.name)

    raise FileNotFoundError(
        "Não encontrei a pasta-base contendo "
        "'Casos-TB-2001-2024', 'DADOS-BASE-MAPA' e 'População'."
    )


def encontrar_pasta(base, nome_esperado):
    """
    Encontra pasta mesmo com problema de acento.
    """
    base = Path(base)

    direto = base / nome_esperado
    if direto.exists():
        return direto

    alvo = normalizar_path_text(nome_esperado)

    for p in base.iterdir():
        if p.is_dir() and normalizar_path_text(p.name) == alvo:
            return p

    raise FileNotFoundError(f"Pasta não encontrada: {nome_esperado}")


def procurar_arquivo(pasta, padroes):
    """
    Procura arquivo dentro de uma pasta usando padrões flexíveis.
    """
    pasta = Path(pasta)

    arquivos = [p for p in pasta.rglob("*") if p.is_file()]

    for arq in arquivos:
        nome = normalizar_path_text(arq.name)

        for padrao in padroes:
            padrao_norm = normalizar_path_text(padrao)

            if re.search(padrao_norm, nome):
                return arq

    print(f"\nArquivos encontrados em {pasta}:")
    for arq in arquivos:
        print(" -", arq.name)

    raise FileNotFoundError(
        f"Nenhum arquivo encontrado em {pasta} para os padrões: {padroes}"
    )


def salvar_fig(nome, pasta_saida):
    """
    Salva e exibe figura.
    """
    path = pasta_saida / nome
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    print("Salvo em:", path)


def validar_colunas(df, colunas):
    ausentes = [c for c in colunas if c not in df.columns]
    if ausentes:
        raise ValueError(f"Colunas ausentes no painel: {ausentes}")


def tendencia_linear(x, y):
    """
    Calcula tendência linear simples.
    Retorna inclinação anual, intercepto e R².
    Uso apenas descritivo.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)

    if mask.sum() < 3:
        return pd.Series({
            "slope_anual": np.nan,
            "intercepto": np.nan,
            "r2": np.nan
        })

    x = x[mask]
    y = y[mask]

    slope, intercept = np.polyfit(x, y, 1)
    yhat = slope * x + intercept

    ss_res = np.sum((y - yhat) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)

    r2 = np.nan if ss_tot == 0 else 1 - ss_res / ss_tot

    return pd.Series({
        "slope_anual": slope,
        "intercepto": intercept,
        "r2": r2
    })


def classificar_porte_pop(pop):
    if pd.isna(pop):
        return np.nan
    if pop < 5_000:
        return "< 5 mil"
    if pop < 10_000:
        return "5–10 mil"
    if pop < 20_000:
        return "10–20 mil"
    if pop < 50_000:
        return "20–50 mil"
    if pop < 100_000:
        return "50–100 mil"
    if pop < 500_000:
        return "100–500 mil"
    return "> 500 mil"


def periodo_quinquenal(ano):
    if 2001 <= ano <= 2005:
        return "2001–2005"
    if 2006 <= ano <= 2010:
        return "2006–2010"
    if 2011 <= ano <= 2015:
        return "2011–2015"
    if 2016 <= ano <= 2019:
        return "2016–2019"
    if 2020 <= ano <= 2024:
        return "2020–2024"
    return np.nan


# ============================================================
# 2. LOCALIZAÇÃO DOS ARQUIVOS
# ============================================================

BASE_DIR = encontrar_base_dir()

DIR_CASOS = encontrar_pasta(BASE_DIR, "Casos-TB-2001-2024")
DIR_GEO = encontrar_pasta(BASE_DIR, "DADOS-BASE-MAPA")
DIR_POP = encontrar_pasta(BASE_DIR, "População")

ARQ_CASOS_SP = DIR_CASOS / "Casos-TB-SP-NOVO.csv"
ARQ_CASOS_RJ = DIR_CASOS / "Casos-TB-RJ-NOVO.csv"

if not ARQ_CASOS_SP.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {ARQ_CASOS_SP}")

if not ARQ_CASOS_RJ.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {ARQ_CASOS_RJ}")

ARQ_POPULACAO = DIR_POP / "PopulaçãoCorrigida_interpolada_CAGR_FINAL.xlsx"

if not ARQ_POPULACAO.exists():
    ARQ_POPULACAO = procurar_arquivo(
        DIR_POP,
        [
            r"populacao.*cagr.*final.*\.xlsx$",
            r"populacao.*interpolada.*\.xlsx$",
            r"populacao.*corrigida.*\.xlsx$",
            r".*\.xlsx$"
        ]
    )

ARQ_GEO_RJ = DIR_GEO / "RJ_geojs-33-mun.json"
ARQ_GEO_SP = DIR_GEO / "SP_geojs-35-mun.json"

if not ARQ_GEO_RJ.exists():
    ARQ_GEO_RJ = procurar_arquivo(
        DIR_GEO,
        [
            r"rj.*geo.*\.json$",
            r"33.*mun.*\.json$",
            r"rj.*\.json$"
        ]
    )

if not ARQ_GEO_SP.exists():
    ARQ_GEO_SP = procurar_arquivo(
        DIR_GEO,
        [
            r"sp.*geo.*\.json$",
            r"35.*mun.*\.json$",
            r"sp.*\.json$"
        ]
    )

OUT_DIR = BASE_DIR / "saida_exploratoria_TB"
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_56 = OUT_DIR / "secao_5_6_analise_descritiva"
OUT_56.mkdir(parents=True, exist_ok=True)

print("Pasta-base:", BASE_DIR)
print("Casos SP:", ARQ_CASOS_SP)
print("Casos RJ:", ARQ_CASOS_RJ)
print("População:", ARQ_POPULACAO)
print("Geo RJ:", ARQ_GEO_RJ)
print("Geo SP:", ARQ_GEO_SP)
print("Saída seção 5.6:", OUT_56)


# ============================================================
# 3. LEITURA DOS CASOS
# ============================================================

def read_csv_casos(path):
    """
    Lê arquivos NOVO de casos.
    Estrutura esperada:
    Município de notificação;2001;2002;...;2024;Total
    """
    last_error = None

    for enc in ["latin1", "cp1252", "utf-8-sig", "utf-8"]:
        try:
            df = pd.read_csv(
                path,
                sep=";",
                encoding=enc,
                dtype=str,
                engine="python",
                quoting=csv.QUOTE_NONE,
                keep_default_na=False,
                on_bad_lines="warn"
            )

            df.columns = [
                str(c).replace('"', "").replace("'", "").strip()
                for c in df.columns
            ]

            # remove colunas vazias
            empty_cols = []
            for c in df.columns:
                vals = df[c].astype(str).str.strip()
                if vals.eq("").all():
                    empty_cols.append(c)

            if empty_cols:
                df = df.drop(columns=empty_cols)

            return df

        except Exception as e:
            last_error = e

    raise last_error


def clean_count_series(s):
    """
    Converte colunas de casos para inteiro.
    """
    s = s.astype(str).str.strip()

    s = (
        s.str.replace('"', "", regex=False)
         .str.replace("'", "", regex=False)
         .str.strip()
    )

    s = s.replace({
        "-": "0",
        "": np.nan,
        "nan": np.nan,
        "NaN": np.nan,
        "None": np.nan
    })

    s = (
        s.str.replace(" ", "", regex=False)
         .str.replace(".", "", regex=False)
         .str.replace(",", ".", regex=False)
    )

    return pd.to_numeric(s, errors="coerce").fillna(0).astype(int)


def carregar_casos(path, estado):
    raw = read_csv_casos(path).copy()

    print(f"\nLendo casos {estado}:")
    print(path)
    print("Colunas encontradas:")
    print(raw.columns.tolist())

    # Identifica coluna de município
    col_mun = None

    for c in raw.columns:
        if "MUNICIPIO" in norm_text(c):
            col_mun = c
            break

    if col_mun is None:
        for c in raw.columns:
            serie = (
                raw[c]
                .astype(str)
                .str.replace('"', "", regex=False)
                .str.replace("'", "", regex=False)
                .str.strip()
            )

            n_match = serie.str.match(r"^\d{6}\s+.+", na=False).sum()

            if n_match >= 5:
                col_mun = c
                break

    if col_mun is None:
        print("\nPrévia do arquivo:")
        display(raw.head(10))
        raise ValueError(f"Não encontrei coluna de município no arquivo: {path}")

    print("Coluna de município identificada:", col_mun)

    raw = raw[raw[col_mun].notna()].copy()

    raw[col_mun] = (
        raw[col_mun]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )

    raw = raw[~raw[col_mun].str.upper().isin(["TOTAL", ""])].copy()

    extracted = raw[col_mun].str.extract(r"^\s*(\d{6})\s+(.+?)\s*$")

    raw["cod_mun6"] = extracted[0]
    raw["municipio"] = extracted[1]

    raw = raw[raw["cod_mun6"].notna()].copy()

    raw["municipio"] = (
        raw["municipio"]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )

    year_cols = []

    for c in raw.columns:
        c_str = str(c).replace('"', "").replace("'", "").strip()

        if c_str.isdigit():
            ano = int(c_str)
            if ano in YEARS:
                year_cols.append(c)

    if len(year_cols) == 0:
        print("\nPrévia do arquivo:")
        display(raw.head(10))
        raise ValueError(f"Não encontrei colunas de anos em: {path}")

    print(f"Anos encontrados em {estado}:")
    print([str(c) for c in year_cols])

    for c in year_cols:
        raw[c] = clean_count_series(raw[c])

    long = raw[["cod_mun6", "municipio"] + year_cols].melt(
        id_vars=["cod_mun6", "municipio"],
        var_name="ano",
        value_name="casos"
    )

    long["ano"] = (
        long["ano"]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .astype(int)
    )

    long["estado"] = estado
    long["municipio"] = long["municipio"].map(norm_text)

    long = long[["estado", "cod_mun6", "municipio", "ano", "casos"]]

    return long


casos_sp = carregar_casos(ARQ_CASOS_SP, "SP")
casos_rj = carregar_casos(ARQ_CASOS_RJ, "RJ")

casos = pd.concat([casos_sp, casos_rj], ignore_index=True)

print("\nCasos carregados:")
display(
    casos.groupby("estado")
    .agg(
        n_linhas=("casos", "size"),
        n_municipios=("cod_mun6", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        casos_total=("casos", "sum")
    )
    .reset_index()
)


# ============================================================
# 4. LEITURA DA POPULAÇÃO
# ============================================================

def clean_pop_series(s):
    """
    Converte população para numérico.
    """
    if pd.api.types.is_numeric_dtype(s):
        return pd.to_numeric(s, errors="coerce").round()

    txt = s.astype(str).str.strip()

    txt = (
        txt.str.replace('"', "", regex=False)
           .str.replace("'", "", regex=False)
           .str.replace(" ", "", regex=False)
    )

    mask_brasil = txt.str.contains(",", regex=False)

    txt_brasil = (
        txt[mask_brasil]
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False)
    )

    txt_outros = txt[~mask_brasil]

    out = pd.Series(index=s.index, dtype="float")
    out.loc[mask_brasil] = pd.to_numeric(txt_brasil, errors="coerce")
    out.loc[~mask_brasil] = pd.to_numeric(txt_outros, errors="coerce")

    return out.round()


def encontrar_sheet_estado(xlsx_path, estado):
    xl = pd.ExcelFile(xlsx_path)
    sheets = xl.sheet_names

    if estado == "RJ":
        termos = ["RJ", "RIO DE JANEIRO"]
    elif estado == "SP":
        termos = ["SP", "SAO PAULO", "SÃO PAULO"]
    else:
        termos = [estado]

    for sh in sheets:
        sh_norm = norm_text(sh)
        if any(norm_text(t) in sh_norm for t in termos):
            return sh

    print("Abas disponíveis:", sheets)
    raise ValueError(f"Não encontrei aba para {estado} em {xlsx_path}")


def identificar_coluna_codigo(df):
    for c in df.columns:
        cn = norm_text(c)
        if "COD" in cn:
            return c

    for c in df.columns:
        serie = df[c].astype(str)
        n_match = serie.str.extract(r"(\d{6})")[0].notna().sum()
        if n_match >= 5:
            return c

    return None


def identificar_coluna_municipio(df):
    for c in df.columns:
        cn = norm_text(c)
        if "MUNICIPIO" in cn:
            return c

    for c in df.columns:
        cn = norm_text(c)
        if cn in ["NOME", "NOME_MUNICIPIO", "NM_MUN", "NM_MUNICIPIO"]:
            return c

    return None


def carregar_populacao(path):
    partes = []

    for estado in ["RJ", "SP"]:
        sheet = encontrar_sheet_estado(path, estado)

        print(f"\nLendo população {estado}:")
        print("Aba:", sheet)

        df = pd.read_excel(path, sheet_name=sheet).copy()

        print("Colunas encontradas:")
        print(df.columns.tolist())

        col_cod = identificar_coluna_codigo(df)
        col_mun = identificar_coluna_municipio(df)

        if col_cod is None:
            raise ValueError(f"Não encontrei coluna de código municipal na aba {sheet}")

        if col_mun is None:
            raise ValueError(f"Não encontrei coluna de município na aba {sheet}")

        df["cod_mun6"] = df[col_cod].astype(str).str.extract(r"(\d{6})")[0]
        df["municipio"] = df[col_mun].map(norm_text)

        year_cols = []

        for c in df.columns:
            c_str = str(c).strip()

            if c_str.isdigit():
                ano = int(c_str)
                if ano in YEARS:
                    year_cols.append(c)

        if len(year_cols) == 0:
            raise ValueError(f"Não encontrei colunas de anos na aba {sheet}")

        print(f"Anos de população encontrados em {estado}:")
        print([str(c) for c in year_cols])

        long = df[["cod_mun6", "municipio"] + year_cols].melt(
            id_vars=["cod_mun6", "municipio"],
            var_name="ano",
            value_name="populacao"
        )

        long["ano"] = long["ano"].astype(int)
        long["populacao"] = clean_pop_series(long["populacao"])
        long["estado"] = estado

        partes.append(
            long[["estado", "cod_mun6", "municipio", "ano", "populacao"]]
        )

    return pd.concat(partes, ignore_index=True)


populacao = carregar_populacao(ARQ_POPULACAO)

print("\nPopulação carregada:")
display(
    populacao.groupby("estado")
    .agg(
        n_linhas=("populacao", "size"),
        n_municipios=("cod_mun6", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        pop_ausente=("populacao", lambda x: x.isna().sum()),
        pop_min=("populacao", "min"),
        pop_max=("populacao", "max")
    )
    .reset_index()
)


# ============================================================
# 5. RECONSTRUÇÃO DO PAINEL MUNICÍPIO-ANO
# ============================================================

painel = casos.merge(
    populacao,
    on=["estado", "cod_mun6", "ano"],
    how="outer",
    suffixes=("_casos", "_pop"),
    indicator=True
)

painel["municipio"] = painel["municipio_casos"].combine_first(painel["municipio_pop"])
painel["casos"] = painel["casos"].fillna(0).astype(int)

painel = painel.sort_values(["estado", "cod_mun6", "ano"]).reset_index(drop=True)

painel["incidencia_100mil"] = np.where(
    painel["populacao"].gt(0),
    painel["casos"] / painel["populacao"] * 100000,
    np.nan
)

painel["log_pop"] = np.where(
    painel["populacao"].gt(0),
    np.log(painel["populacao"]),
    np.nan
)

painel["var_pop_pct"] = (
    painel
    .groupby(["estado", "cod_mun6"])["populacao"]
    .pct_change() * 100
)

painel = painel[
    [
        "estado",
        "cod_mun6",
        "municipio",
        "ano",
        "casos",
        "populacao",
        "log_pop",
        "incidencia_100mil",
        "var_pop_pct",
        "_merge"
    ]
]

print("\nPainel reconstruído:")
display(
    painel.groupby("estado")
    .agg(
        n_linhas=("casos", "size"),
        n_municipios=("cod_mun6", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        casos_total=("casos", "sum"),
        pop_min=("populacao", "min"),
        pop_max=("populacao", "max")
    )
    .reset_index()
)

print("\nDiagnóstico do merge casos × população:")
display(painel["_merge"].value_counts().to_frame("n"))

painel.to_csv(
    OUT_56 / "painel_reconstruido_TB_SP_RJ_2001_2024.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 6. CARREGAMENTO DAS MALHAS, SE GEOPANDAS ESTIVER DISPONÍVEL
# ============================================================

geo_sp = None
geo_rj = None

try:
    import geopandas as gpd

    def carregar_malha(path, estado):
        gdf = gpd.read_file(path).copy()

        if "id" not in gdf.columns:
            raise ValueError(f"A malha {path} não tem coluna 'id'.")

        gdf["cod_mun6"] = gdf["id"].astype(str).str[:6]

        if "name" in gdf.columns:
            gdf["municipio_geo"] = gdf["name"].map(norm_text)
        else:
            gdf["municipio_geo"] = ""

        gdf["estado"] = estado

        if gdf.crs is None:
            gdf = gdf.set_crs("EPSG:4326")

        return gdf

    geo_rj = carregar_malha(ARQ_GEO_RJ, "RJ")
    geo_sp = carregar_malha(ARQ_GEO_SP, "SP")

    print("\nMalhas carregadas:")
    display(
        pd.DataFrame({
            "estado": ["RJ", "SP"],
            "municipios_malha": [
                geo_rj["cod_mun6"].nunique(),
                geo_sp["cod_mun6"].nunique()
            ]
        })
    )

except Exception as e:
    print("\nMalhas não foram carregadas. Os mapas serão ignorados.")
    print("Motivo:", e)
    print("Para mapas, instale:")
    print("!pip install geopandas mapclassify")


# ============================================================
# 7. PREPARAÇÃO DO PAINEL PARA A SEÇÃO
# ============================================================

validar_colunas(
    painel,
    [COL_ESTADO, COL_COD, COL_MUN, COL_ANO, COL_CASOS, COL_POP, COL_INC]
)

df56 = painel.copy()

df56[COL_ANO] = df56[COL_ANO].astype(int)
df56[COL_CASOS] = pd.to_numeric(df56[COL_CASOS], errors="coerce")
df56[COL_POP] = pd.to_numeric(df56[COL_POP], errors="coerce")
df56[COL_INC] = pd.to_numeric(df56[COL_INC], errors="coerce")

df56 = df56[df56[COL_ANO].isin(YEARS)].copy()

df56["caso_zero"] = df56[COL_CASOS].eq(0)
df56["caso_positivo"] = df56[COL_CASOS].gt(0)
df56["porte_pop"] = df56[COL_POP].apply(classificar_porte_pop)
df56["periodo"] = df56[COL_ANO].apply(periodo_quinquenal)
df56["log_casos_mais1"] = np.log1p(df56[COL_CASOS])
df56["log_inc_mais1"] = np.log1p(df56[COL_INC])

print("\nPainel da seção 5.6 preparado:")
display(df56.head())


# ============================================================
# 8. AUDITORIA DESCRITIVA DO PAINEL
# ============================================================

municipios_esperados = {
    "SP": 645,
    "RJ": 92
}

auditoria_56_estado = (
    df56
    .groupby(COL_ESTADO)
    .agg(
        linhas_municipio_ano=(COL_CASOS, "size"),
        municipios=(COL_COD, "nunique"),
        anos=(COL_ANO, "nunique"),
        ano_min=(COL_ANO, "min"),
        ano_max=(COL_ANO, "max"),
        casos_total=(COL_CASOS, "sum"),
        populacao_ausente=(COL_POP, lambda x: x.isna().sum()),
        populacao_zero_ou_negativa=(COL_POP, lambda x: x.le(0).sum()),
        incidencia_ausente=(COL_INC, lambda x: x.isna().sum()),
        casos_negativos=(COL_CASOS, lambda x: x.lt(0).sum())
    )
    .reset_index()
)

auditoria_56_estado["municipios_esperados"] = auditoria_56_estado[COL_ESTADO].map(municipios_esperados)
auditoria_56_estado["diferenca_municipios"] = (
    auditoria_56_estado["municipios"] -
    auditoria_56_estado["municipios_esperados"]
)

linhas_por_municipio = (
    df56
    .groupby([COL_ESTADO, COL_COD, COL_MUN])
    .agg(
        n_linhas=(COL_ANO, "size"),
        anos_distintos=(COL_ANO, "nunique"),
        ano_min=(COL_ANO, "min"),
        ano_max=(COL_ANO, "max")
    )
    .reset_index()
)

municipios_painel_incompleto = linhas_por_municipio[
    (linhas_por_municipio["n_linhas"] != len(YEARS)) |
    (linhas_por_municipio["anos_distintos"] != len(YEARS))
].copy()

print("\nAuditoria descritiva por estado:")
display(auditoria_56_estado)

print("\nMunicípios com painel incompleto:")
display(municipios_painel_incompleto)


# ============================================================
# 9. INDICADORES EPIDEMIOLÓGICOS ESTADO-ANO
# ============================================================

estado_ano_56 = (
    df56
    .groupby([COL_ESTADO, COL_ANO])
    .agg(
        casos_total=(COL_CASOS, "sum"),
        populacao_total=(COL_POP, "sum"),
        municipios=(COL_COD, "nunique"),
        municipios_zero=("caso_zero", "sum"),
        media_casos_municipio=(COL_CASOS, "mean"),
        mediana_casos_municipio=(COL_CASOS, "median"),
        variancia_casos_municipio=(COL_CASOS, "var"),
        media_incidencia_municipal=(COL_INC, "mean"),
        mediana_incidencia_municipal=(COL_INC, "median"),
        p75_incidencia_municipal=(COL_INC, lambda x: x.quantile(0.75)),
        p90_incidencia_municipal=(COL_INC, lambda x: x.quantile(0.90)),
        p95_incidencia_municipal=(COL_INC, lambda x: x.quantile(0.95)),
        p99_incidencia_municipal=(COL_INC, lambda x: x.quantile(0.99)),
        max_incidencia_municipal=(COL_INC, "max")
    )
    .reset_index()
)

estado_ano_56["incidencia_estado_100mil"] = (
    estado_ano_56["casos_total"] /
    estado_ano_56["populacao_total"] * 100000
)

estado_ano_56["prop_municipios_zero"] = (
    estado_ano_56["municipios_zero"] /
    estado_ano_56["municipios"]
)

estado_ano_56["razao_var_media_casos"] = (
    estado_ano_56["variancia_casos_municipio"] /
    estado_ano_56["media_casos_municipio"].replace(0, np.nan)
)

print("\nIndicadores estado-ano:")
display(estado_ano_56.head())


# ============================================================
# 10. INDICADORES MUNICIPAIS DO PERÍODO 2001–2024
# ============================================================

municipio_periodo_56 = (
    df56
    .groupby([COL_ESTADO, COL_COD, COL_MUN])
    .agg(
        casos_acumulados=(COL_CASOS, "sum"),
        casos_media_anual=(COL_CASOS, "mean"),
        casos_mediana_anual=(COL_CASOS, "median"),
        casos_max_anual=(COL_CASOS, "max"),
        populacao_media=(COL_POP, "mean"),
        populacao_min=(COL_POP, "min"),
        populacao_max=(COL_POP, "max"),
        incidencia_media_anual=(COL_INC, "mean"),
        incidencia_mediana_anual=(COL_INC, "median"),
        incidencia_max_anual=(COL_INC, "max"),
        anos_com_casos=("caso_positivo", "sum"),
        anos_zero=("caso_zero", "sum")
    )
    .reset_index()
)

pop_pessoa_ano = (
    df56
    .groupby([COL_ESTADO, COL_COD, COL_MUN])[COL_POP]
    .sum()
    .reset_index(name="pop_pessoa_ano")
)

municipio_periodo_56 = municipio_periodo_56.merge(
    pop_pessoa_ano,
    on=[COL_ESTADO, COL_COD, COL_MUN],
    how="left"
)

municipio_periodo_56["incidencia_periodo_100mil_pessoa_ano"] = (
    municipio_periodo_56["casos_acumulados"] /
    municipio_periodo_56["pop_pessoa_ano"] * 100000
)

municipio_periodo_56["prop_anos_zero"] = (
    municipio_periodo_56["anos_zero"] / len(YEARS)
)

municipio_periodo_56["porte_pop_medio"] = (
    municipio_periodo_56["populacao_media"].apply(classificar_porte_pop)
)

municipio_periodo_56 = municipio_periodo_56.sort_values(
    [COL_ESTADO, "casos_acumulados"],
    ascending=[True, False]
)

municipio_periodo_56["rank_casos_estado"] = (
    municipio_periodo_56
    .groupby(COL_ESTADO)["casos_acumulados"]
    .rank(method="first", ascending=False)
    .astype(int)
)

municipio_periodo_56["rank_incidencia_estado"] = (
    municipio_periodo_56
    .groupby(COL_ESTADO)["incidencia_periodo_100mil_pessoa_ano"]
    .rank(method="first", ascending=False)
    .astype(int)
)

print("\nIndicadores municipais do período:")
display(municipio_periodo_56.head(20))


# ============================================================
# 11. MEDIDAS-RESUMO POR ESTADO
# ============================================================

resumo_estado_56 = (
    municipio_periodo_56
    .groupby(COL_ESTADO)
    .agg(
        municipios=(COL_COD, "nunique"),
        casos_acumulados_total=("casos_acumulados", "sum"),
        media_casos_acumulados_mun=("casos_acumulados", "mean"),
        mediana_casos_acumulados_mun=("casos_acumulados", "median"),
        p75_casos_acumulados_mun=("casos_acumulados", lambda x: x.quantile(0.75)),
        p90_casos_acumulados_mun=("casos_acumulados", lambda x: x.quantile(0.90)),
        p95_casos_acumulados_mun=("casos_acumulados", lambda x: x.quantile(0.95)),
        max_casos_acumulados_mun=("casos_acumulados", "max"),
        media_incidencia_periodo_mun=("incidencia_periodo_100mil_pessoa_ano", "mean"),
        mediana_incidencia_periodo_mun=("incidencia_periodo_100mil_pessoa_ano", "median"),
        p75_incidencia_periodo_mun=("incidencia_periodo_100mil_pessoa_ano", lambda x: x.quantile(0.75)),
        p90_incidencia_periodo_mun=("incidencia_periodo_100mil_pessoa_ano", lambda x: x.quantile(0.90)),
        p95_incidencia_periodo_mun=("incidencia_periodo_100mil_pessoa_ano", lambda x: x.quantile(0.95)),
        max_incidencia_periodo_mun=("incidencia_periodo_100mil_pessoa_ano", "max"),
        media_anos_zero=("anos_zero", "mean"),
        mediana_anos_zero=("anos_zero", "median")
    )
    .reset_index()
)

estado_periodo_total = (
    df56
    .groupby(COL_ESTADO)
    .agg(
        casos_total=(COL_CASOS, "sum"),
        pop_pessoa_ano=(COL_POP, "sum")
    )
    .reset_index()
)

estado_periodo_total["incidencia_periodo_estado_100mil_pessoa_ano"] = (
    estado_periodo_total["casos_total"] /
    estado_periodo_total["pop_pessoa_ano"] * 100000
)

resumo_estado_56 = resumo_estado_56.merge(
    estado_periodo_total[
        [COL_ESTADO, "incidencia_periodo_estado_100mil_pessoa_ano"]
    ],
    on=COL_ESTADO,
    how="left"
)

print("\nMedidas-resumo por estado:")
display(resumo_estado_56)


# ============================================================
# 12. COMPARAÇÃO EPIDEMIOLÓGICA INICIAL RJ vs SP
# ============================================================

comp_wide = resumo_estado_56.set_index(COL_ESTADO)

linhas_comparacao = []

if "RJ" in comp_wide.index and "SP" in comp_wide.index:
    variaveis_comp = [
        "casos_acumulados_total",
        "incidencia_periodo_estado_100mil_pessoa_ano",
        "media_incidencia_periodo_mun",
        "mediana_incidencia_periodo_mun",
        "p90_incidencia_periodo_mun",
        "max_incidencia_periodo_mun",
        "media_anos_zero"
    ]

    for var in variaveis_comp:
        rj = comp_wide.loc["RJ", var]
        sp = comp_wide.loc["SP", var]

        linhas_comparacao.append({
            "indicador": var,
            "RJ": rj,
            "SP": sp,
            "diferenca_RJ_menos_SP": rj - sp,
            "razao_RJ_por_SP": rj / sp if pd.notna(sp) and sp != 0 else np.nan
        })

comparacao_rj_sp_56 = pd.DataFrame(linhas_comparacao)

print("\nComparação epidemiológica inicial RJ vs SP:")
display(comparacao_rj_sp_56)


# ============================================================
# 13. TENDÊNCIAS TEMPORAIS OBSERVADAS POR ESTADO
# ============================================================

tendencia_estado_casos = (
    estado_ano_56
    .groupby(COL_ESTADO)
    .apply(lambda d: tendencia_linear(d[COL_ANO], d["casos_total"]))
    .reset_index()
    .rename(columns={
        "slope_anual": "slope_anual_casos",
        "intercepto": "intercepto_casos",
        "r2": "r2_casos"
    })
)

tendencia_estado_inc = (
    estado_ano_56
    .groupby(COL_ESTADO)
    .apply(lambda d: tendencia_linear(d[COL_ANO], d["incidencia_estado_100mil"]))
    .reset_index()
    .rename(columns={
        "slope_anual": "slope_anual_incidencia",
        "intercepto": "intercepto_incidencia",
        "r2": "r2_incidencia"
    })
)

tendencia_estado_56 = tendencia_estado_casos.merge(
    tendencia_estado_inc,
    on=COL_ESTADO,
    how="left"
)

print("\nTendências temporais observadas por estado:")
display(tendencia_estado_56)


# ============================================================
# 14. ALTA INCIDÊNCIA PERSISTENTE
# ============================================================

limiares_estado_ano_56 = (
    df56
    .groupby([COL_ESTADO, COL_ANO])
    .agg(
        p90_inc_estado_ano=(COL_INC, lambda x: x.quantile(0.90)),
        p95_inc_estado_ano=(COL_INC, lambda x: x.quantile(0.95))
    )
    .reset_index()
)

df56 = df56.merge(
    limiares_estado_ano_56,
    on=[COL_ESTADO, COL_ANO],
    how="left"
)

df56["acima_p90_inc_estado_ano"] = (
    df56[COL_INC] >= df56["p90_inc_estado_ano"]
)

df56["acima_p95_inc_estado_ano"] = (
    df56[COL_INC] >= df56["p95_inc_estado_ano"]
)

persistencia_56 = (
    df56
    .groupby([COL_ESTADO, COL_COD, COL_MUN])
    .agg(
        casos_acumulados=(COL_CASOS, "sum"),
        incidencia_periodo_media=(COL_INC, "mean"),
        incidencia_max=(COL_INC, "max"),
        anos_acima_p90=("acima_p90_inc_estado_ano", "sum"),
        anos_acima_p95=("acima_p95_inc_estado_ano", "sum"),
        anos_com_casos=("caso_positivo", "sum"),
        anos_zero=("caso_zero", "sum"),
        populacao_media=(COL_POP, "mean")
    )
    .reset_index()
)

persistencia_56["prop_anos_acima_p90"] = (
    persistencia_56["anos_acima_p90"] / len(YEARS)
)

persistencia_56["prop_anos_acima_p95"] = (
    persistencia_56["anos_acima_p95"] / len(YEARS)
)

persistencia_56["alta_incidencia_persistente_p90"] = (
    persistencia_56["anos_acima_p90"] >= 5
)

persistencia_56["alta_incidencia_persistente_p95"] = (
    persistencia_56["anos_acima_p95"] >= 5
)

print("\nMunicípios com alta incidência persistente — p95, pelo menos 5 anos:")
display(
    persistencia_56
    .query("alta_incidencia_persistente_p95 == True")
    .sort_values([COL_ESTADO, "anos_acima_p95", "casos_acumulados"], ascending=[True, False, False])
    .groupby(COL_ESTADO)
    .head(20)
)


# ============================================================
# 15. CONCENTRAÇÃO DOS CASOS
# ============================================================

concentracao_base = municipio_periodo_56.copy()

total_casos_estado = (
    concentracao_base
    .groupby(COL_ESTADO)["casos_acumulados"]
    .transform("sum")
)

concentracao_base["perc_casos_estado"] = (
    concentracao_base["casos_acumulados"] /
    total_casos_estado * 100
)

concentracao_base = concentracao_base.sort_values(
    [COL_ESTADO, "casos_acumulados"],
    ascending=[True, False]
)

concentracao_base["rank_casos_estado"] = (
    concentracao_base
    .groupby(COL_ESTADO)["casos_acumulados"]
    .rank(method="first", ascending=False)
    .astype(int)
)

concentracao_base["perc_acumulado_casos_estado"] = (
    concentracao_base
    .groupby(COL_ESTADO)["perc_casos_estado"]
    .cumsum()
)

limiares_concentracao = [50, 70, 80, 90]
linhas_concentracao = []

for estado, sub in concentracao_base.groupby(COL_ESTADO):
    sub = sub.sort_values("rank_casos_estado")

    for limiar in limiares_concentracao:
        if (sub["perc_acumulado_casos_estado"] >= limiar).any():
            n_mun = int(
                sub.loc[
                    sub["perc_acumulado_casos_estado"] >= limiar,
                    "rank_casos_estado"
                ].min()
            )
        else:
            n_mun = sub[COL_COD].nunique()

        linhas_concentracao.append({
            COL_ESTADO: estado,
            "limiar_percentual_casos": limiar,
            "n_municipios_necessarios": n_mun,
            "total_municipios_estado": sub[COL_COD].nunique(),
            "perc_municipios_necessarios": n_mun / sub[COL_COD].nunique() * 100
        })

concentracao_56 = pd.DataFrame(linhas_concentracao)

print("\nConcentração dos casos acumulados:")
display(concentracao_56)


# ============================================================
# 16. INDICADORES POR PORTE POPULACIONAL
# ============================================================

ordem_porte = [
    "< 5 mil",
    "5–10 mil",
    "10–20 mil",
    "20–50 mil",
    "50–100 mil",
    "100–500 mil",
    "> 500 mil"
]

porte_56 = (
    df56
    .groupby([COL_ESTADO, "porte_pop"])
    .agg(
        linhas_municipio_ano=(COL_CASOS, "size"),
        municipios=(COL_COD, "nunique"),
        casos_total=(COL_CASOS, "sum"),
        casos_media_municipio_ano=(COL_CASOS, "mean"),
        incidencia_media=(COL_INC, "mean"),
        incidencia_mediana=(COL_INC, "median"),
        incidencia_p90=(COL_INC, lambda x: x.quantile(0.90)),
        incidencia_p95=(COL_INC, lambda x: x.quantile(0.95)),
        incidencia_max=(COL_INC, "max"),
        prop_zeros=("caso_zero", "mean")
    )
    .reset_index()
)

porte_56["porte_pop"] = pd.Categorical(
    porte_56["porte_pop"],
    categories=ordem_porte,
    ordered=True
)

porte_56 = porte_56.sort_values([COL_ESTADO, "porte_pop"])

print("\nIndicadores por porte populacional:")
display(porte_56)


# ============================================================
# 17. INDICADORES POR PERÍODO
# ============================================================

periodo_estado_56 = (
    df56
    .dropna(subset=["periodo"])
    .groupby([COL_ESTADO, "periodo"])
    .agg(
        casos_total=(COL_CASOS, "sum"),
        populacao_pessoa_ano=(COL_POP, "sum"),
        municipios=(COL_COD, "nunique"),
        media_inc_municipal=(COL_INC, "mean"),
        mediana_inc_municipal=(COL_INC, "median"),
        prop_zeros=("caso_zero", "mean")
    )
    .reset_index()
)

periodo_estado_56["incidencia_periodo_estado_100mil_pessoa_ano"] = (
    periodo_estado_56["casos_total"] /
    periodo_estado_56["populacao_pessoa_ano"] * 100000
)

periodo_municipio_56 = (
    df56
    .dropna(subset=["periodo"])
    .groupby([COL_ESTADO, COL_COD, COL_MUN, "periodo"])
    .agg(
        casos_total=(COL_CASOS, "sum"),
        populacao_pessoa_ano=(COL_POP, "sum"),
        incidencia_media_anual=(COL_INC, "mean"),
        anos=(COL_ANO, "nunique")
    )
    .reset_index()
)

periodo_municipio_56["incidencia_periodo_100mil_pessoa_ano"] = (
    periodo_municipio_56["casos_total"] /
    periodo_municipio_56["populacao_pessoa_ano"] * 100000
)

print("\nIndicadores por período — estado:")
display(periodo_estado_56)


# ============================================================
# 18. FIGURAS — SÉRIES TEMPORAIS ESTADUAIS
# ============================================================

plt.figure(figsize=(10, 5))

for estado, sub in estado_ano_56.groupby(COL_ESTADO):
    sub = sub.sort_values(COL_ANO)
    plt.plot(
        sub[COL_ANO],
        sub["casos_total"],
        marker="o",
        linewidth=2,
        label=estado
    )

plt.title("Casos totais de tuberculose por ano — SP e RJ")
plt.xlabel("Ano")
plt.ylabel("Casos")
plt.grid(alpha=0.3)
plt.legend()
salvar_fig("serie_temporal_casos_estado.png", OUT_56)


plt.figure(figsize=(10, 5))

for estado, sub in estado_ano_56.groupby(COL_ESTADO):
    sub = sub.sort_values(COL_ANO)
    plt.plot(
        sub[COL_ANO],
        sub["incidencia_estado_100mil"],
        marker="o",
        linewidth=2,
        label=estado
    )

plt.title("Incidência anual de tuberculose por 100 mil habitantes — SP e RJ")
plt.xlabel("Ano")
plt.ylabel("Incidência por 100 mil")
plt.grid(alpha=0.3)
plt.legend()
salvar_fig("serie_temporal_incidencia_estado.png", OUT_56)


plt.figure(figsize=(10, 5))

for estado, sub in estado_ano_56.groupby(COL_ESTADO):
    sub = sub.sort_values(COL_ANO)
    plt.plot(
        sub[COL_ANO],
        sub["prop_municipios_zero"],
        marker="o",
        linewidth=2,
        label=estado
    )

plt.title("Proporção de municípios com zero casos por ano — SP e RJ")
plt.xlabel("Ano")
plt.ylabel("Proporção de municípios com zero casos")
plt.grid(alpha=0.3)
plt.legend()
salvar_fig("serie_temporal_prop_municipios_zero.png", OUT_56)


# ============================================================
# 19. FIGURAS — DISTRIBUIÇÕES MUNICIPAIS
# ============================================================

for estado in ["RJ", "SP"]:
    sub = df56[df56[COL_ESTADO] == estado].copy()

    limite = sub[COL_INC].quantile(0.99)
    x = sub[COL_INC].clip(upper=limite)

    plt.figure(figsize=(9, 5))
    plt.hist(x.dropna(), bins=50)
    plt.title(
        f"Distribuição da incidência município-ano — {estado}\n"
        f"valores acima do p99 truncados para visualização"
    )
    plt.xlabel("Incidência por 100 mil")
    plt.ylabel("Frequência município-ano")
    plt.grid(alpha=0.3)
    salvar_fig(f"hist_incidencia_municipio_ano_{estado}.png", OUT_56)


for estado in ["RJ", "SP"]:
    sub = df56[df56[COL_ESTADO] == estado].copy()

    limite = sub[COL_CASOS].quantile(0.99)
    x = sub[COL_CASOS].clip(upper=limite)

    plt.figure(figsize=(9, 5))
    plt.hist(x.dropna(), bins=50)
    plt.title(
        f"Distribuição dos casos município-ano — {estado}\n"
        f"valores acima do p99 truncados para visualização"
    )
    plt.xlabel("Casos por município-ano")
    plt.ylabel("Frequência município-ano")
    plt.grid(alpha=0.3)
    salvar_fig(f"hist_casos_municipio_ano_{estado}.png", OUT_56)


plt.figure(figsize=(8, 5))
df56.boxplot(
    column=COL_INC,
    by=COL_ESTADO,
    showfliers=False
)
plt.title("Distribuição da incidência município-ano por estado")
plt.suptitle("")
plt.xlabel("Estado")
plt.ylabel("Incidência por 100 mil")
plt.grid(alpha=0.3)
salvar_fig("boxplot_incidencia_estado_sem_outliers.png", OUT_56)


plt.figure(figsize=(8, 5))
df56.boxplot(
    column=COL_CASOS,
    by=COL_ESTADO,
    showfliers=False
)
plt.title("Distribuição dos casos município-ano por estado")
plt.suptitle("")
plt.xlabel("Estado")
plt.ylabel("Casos")
plt.grid(alpha=0.3)
salvar_fig("boxplot_casos_estado_sem_outliers.png", OUT_56)


# ============================================================
# 20. FIGURAS — CONCENTRAÇÃO DOS CASOS
# ============================================================

for estado, sub in concentracao_base.groupby(COL_ESTADO):
    sub = sub.sort_values("rank_casos_estado")

    plt.figure(figsize=(10, 5))
    plt.plot(
        sub["rank_casos_estado"],
        sub["perc_acumulado_casos_estado"],
        marker="o",
        markersize=3
    )

    plt.axhline(50, linestyle="--")
    plt.axhline(80, linestyle="--")
    plt.axhline(90, linestyle="--")

    plt.title(f"Curva de concentração dos casos acumulados — {estado}, 2001–2024")
    plt.xlabel("Municípios ordenados por casos acumulados")
    plt.ylabel("% acumulado dos casos do estado")
    plt.grid(alpha=0.3)
    salvar_fig(f"curva_concentracao_casos_{estado}.png", OUT_56)


# ============================================================
# 21. TOP 10 MUNICÍPIOS — SÉRIE ANUAL E ACUMULADA POR ESTADO
# ============================================================

top10_casos_estado_56 = (
    municipio_periodo_56
    .query("rank_casos_estado <= 10")
    .copy()
    .sort_values([COL_ESTADO, "rank_casos_estado"])
)

print("\nTop 10 municípios por casos acumulados em cada estado:")
display(top10_casos_estado_56)


def plot_top10_estado(estado, acumulado=True):
    top = top10_casos_estado_56[top10_casos_estado_56[COL_ESTADO] == estado].copy()

    d = df56.merge(
        top[[COL_ESTADO, COL_COD, COL_MUN, "rank_casos_estado"]],
        on=[COL_ESTADO, COL_COD, COL_MUN],
        how="inner"
    )

    d = d.sort_values(["rank_casos_estado", COL_ANO])

    if acumulado:
        d["valor_plot"] = d.groupby([COL_ESTADO, COL_COD])[COL_CASOS].cumsum()
        titulo_tipo = "casos acumulados"
        ylabel = "Casos acumulados até o ano"
        nome = f"top10_serie_casos_acumulados_{estado}.png"
    else:
        d["valor_plot"] = d[COL_CASOS]
        titulo_tipo = "casos anuais"
        ylabel = "Casos no ano"
        nome = f"top10_serie_casos_anuais_{estado}.png"

    plt.figure(figsize=(13, 7))

    for municipio, sub in d.groupby(COL_MUN):
        sub = sub.sort_values(COL_ANO)
        plt.plot(
            sub[COL_ANO],
            sub["valor_plot"],
            marker="o",
            linewidth=2,
            label=municipio
        )

    plt.title(
        f"Série temporal dos {titulo_tipo} de tuberculose\n"
        f"Top 10 municípios por casos acumulados — {estado}, 2001–2024"
    )
    plt.xlabel("Ano")
    plt.ylabel(ylabel)
    plt.grid(alpha=0.3)
    plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
    salvar_fig(nome, OUT_56)

    return d


serie_top10_rj_acum_56 = plot_top10_estado("RJ", acumulado=True)
serie_top10_sp_acum_56 = plot_top10_estado("SP", acumulado=True)

serie_top10_rj_anual_56 = plot_top10_estado("RJ", acumulado=False)
serie_top10_sp_anual_56 = plot_top10_estado("SP", acumulado=False)


# ============================================================
# 22. MAPAS TEMÁTICOS
# ============================================================

if geo_sp is not None and geo_rj is not None:
    try:
        geo_sp_56 = geo_sp.copy()
        geo_rj_56 = geo_rj.copy()

        geo_rj_56 = geo_rj_56.to_crs(geo_sp_56.crs)

        geo_56 = pd.concat([geo_sp_56, geo_rj_56], ignore_index=True)

        def mapa_municipal(dados, variavel, titulo, nome, log1p=False):
            d = dados.copy()

            variavel_plot = variavel

            if log1p:
                variavel_plot = f"{variavel}_log1p"
                d[variavel_plot] = np.log1p(d[variavel])

            gdf = geo_56.merge(
                d,
                on=[COL_ESTADO, COL_COD],
                how="left"
            )

            fig, ax = plt.subplots(figsize=(12, 10))

            gdf.plot(
                column=variavel_plot,
                ax=ax,
                legend=True,
                linewidth=0.15,
                edgecolor="black",
                missing_kwds={
                    "color": "lightgrey",
                    "label": "Sem dado"
                }
            )

            ax.set_title(titulo, fontsize=14)
            ax.axis("off")

            plt.savefig(OUT_56 / nome, dpi=300, bbox_inches="tight")
            plt.show()
            print("Mapa salvo em:", OUT_56 / nome)

        mapa_municipal(
            municipio_periodo_56,
            "casos_acumulados",
            "Casos acumulados de tuberculose por município\nSP e RJ, 2001–2024",
            "mapa_casos_acumulados_SP_RJ_linear.png",
            log1p=False
        )

        mapa_municipal(
            municipio_periodo_56,
            "casos_acumulados",
            "Casos acumulados de tuberculose por município — log(1 + casos)\nSP e RJ, 2001–2024",
            "mapa_casos_acumulados_SP_RJ_log1p.png",
            log1p=True
        )

        mapa_municipal(
            municipio_periodo_56,
            "incidencia_periodo_100mil_pessoa_ano",
            "Incidência média do período por 100 mil pessoa-ano\nSP e RJ, 2001–2024",
            "mapa_incidencia_periodo_SP_RJ.png",
            log1p=False
        )

        mapa_municipal(
            persistencia_56,
            "anos_acima_p95",
            "Persistência de alta incidência: anos acima do p95 estadual anual\nSP e RJ, 2001–2024",
            "mapa_persistencia_alta_incidencia_p95_SP_RJ.png",
            log1p=False
        )

        for periodo in periodo_municipio_56["periodo"].dropna().unique():
            dper = periodo_municipio_56[periodo_municipio_56["periodo"] == periodo].copy()

            nome_periodo = (
                periodo
                .replace("–", "_")
                .replace("-", "_")
                .replace(" ", "_")
            )

            mapa_municipal(
                dper,
                "incidencia_periodo_100mil_pessoa_ano",
                f"Incidência por 100 mil pessoa-ano — {periodo}\nSP e RJ",
                f"mapa_incidencia_periodo_{nome_periodo}.png",
                log1p=False
            )

    except Exception as e:
        print("\nMapas temáticos não foram gerados.")
        print("Motivo:", e)
else:
    print("\nMapas ignorados porque as malhas não foram carregadas.")


# ============================================================
# 23. EXPORTAÇÃO DOS RESULTADOS DA SEÇÃO 5.6
# ============================================================

tabelas = {
    "auditoria_56_estado": auditoria_56_estado,
    "municipios_painel_incompleto": municipios_painel_incompleto,
    "estado_ano_56": estado_ano_56,
    "municipio_periodo_56": municipio_periodo_56,
    "resumo_estado_56": resumo_estado_56,
    "comparacao_rj_sp_56": comparacao_rj_sp_56,
    "tendencia_estado_56": tendencia_estado_56,
    "persistencia_56": persistencia_56,
    "concentracao_56": concentracao_56,
    "porte_56": porte_56,
    "periodo_estado_56": periodo_estado_56,
    "periodo_municipio_56": periodo_municipio_56,
    "top10_casos_estado_56": top10_casos_estado_56,
    "serie_top10_rj_acum_56": serie_top10_rj_acum_56,
    "serie_top10_sp_acum_56": serie_top10_sp_acum_56,
    "serie_top10_rj_anual_56": serie_top10_rj_anual_56,
    "serie_top10_sp_anual_56": serie_top10_sp_anual_56
}

for nome, tabela in tabelas.items():
    tabela.to_csv(
        OUT_56 / f"{nome}.csv",
        index=False,
        encoding="utf-8-sig"
    )

excel_56 = OUT_56 / "secao_5_6_analise_descritiva_comparacao_epidemiologica.xlsx"

with pd.ExcelWriter(excel_56) as writer:
    auditoria_56_estado.to_excel(writer, sheet_name="auditoria_estado", index=False)
    municipios_painel_incompleto.to_excel(writer, sheet_name="painel_incompleto", index=False)
    estado_ano_56.to_excel(writer, sheet_name="estado_ano", index=False)
    municipio_periodo_56.to_excel(writer, sheet_name="municipio_periodo", index=False)
    resumo_estado_56.to_excel(writer, sheet_name="resumo_estado", index=False)
    comparacao_rj_sp_56.to_excel(writer, sheet_name="comparacao_RJ_SP", index=False)
    tendencia_estado_56.to_excel(writer, sheet_name="tendencias_estado", index=False)
    persistencia_56.to_excel(writer, sheet_name="persistencia", index=False)
    concentracao_56.to_excel(writer, sheet_name="concentracao", index=False)
    porte_56.to_excel(writer, sheet_name="porte_pop", index=False)
    periodo_estado_56.to_excel(writer, sheet_name="periodo_estado", index=False)
    periodo_municipio_56.to_excel(writer, sheet_name="periodo_municipio", index=False)
    top10_casos_estado_56.to_excel(writer, sheet_name="top10_estado", index=False)

print("\n============================================================")
print("SEÇÃO 5.6 FINALIZADA")
print("Painel reconstruído e análise descritiva concluída.")
print("Resultados salvos em:")
print(OUT_56)
print("\nArquivo Excel principal:")
print(excel_56)
print("============================================================")
```
