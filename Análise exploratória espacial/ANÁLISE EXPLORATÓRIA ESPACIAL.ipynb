# ANÁLISE EXPLORATÓRIA ESPACIAL

```python
# ============================================================
# SEÇÃO — ANÁLISE EXPLORATÓRIA ESPACIAL
# Tuberculose — SP e RJ, 2001–2024
#
# Objetivos:
# - reconstruir painel município-ano;
# - carregar malhas municipais;
# - construir matriz de vizinhança queen por estado;
# - calcular Moran Global;
# - calcular LISA;
# - calcular Getis-Ord Gi*;
# - gerar mapas e tabelas diagnósticas.
# ============================================================

from pathlib import Path
import re
import csv
import unicodedata
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

warnings.filterwarnings("ignore")

np.random.seed(12345)

# ============================================================
# 0. CONFIGURAÇÕES GERAIS
# ============================================================

YEARS = list(range(2001, 2025))
ANOS_SELECIONADOS = [2001, 2005, 2010, 2015, 2019, 2020, 2024]

PERMUTATIONS = 999
ALPHA = 0.05

COL_ESTADO = "estado"
COL_COD = "cod_mun6"
COL_MUN = "municipio"
COL_ANO = "ano"
COL_CASOS = "casos"
COL_POP = "populacao"
COL_INC = "incidencia_100mil"


# ============================================================
# 1. FUNÇÕES GERAIS
# ============================================================

def norm_text(x):
    if pd.isna(x):
        return ""

    x = str(x).strip()
    x = unicodedata.normalize("NFKD", x)
    x = "".join(ch for ch in x if not unicodedata.combining(ch))
    x = re.sub(r"\s+", " ", x)

    return x.upper().strip()


def normalizar_path_text(x):
    x = str(x)
    x = unicodedata.normalize("NFKD", x)
    x = "".join(ch for ch in x if not unicodedata.combining(ch))
    return x.lower()


def encontrar_base_dir():
    """
    Procura a pasta-base subindo a partir do diretório atual.
    Espera encontrar:
    - Casos-TB-2001-2024
    - DADOS-BASE-MAPA
    - População
    """
    atual = Path.cwd()
    candidatos = [atual] + list(atual.parents)

    for base in candidatos:
        try:
            nomes = [normalizar_path_text(p.name) for p in base.iterdir() if p.is_dir()]
        except Exception:
            continue

        tem_casos = normalizar_path_text("Casos-TB-2001-2024") in nomes
        tem_mapa = normalizar_path_text("DADOS-BASE-MAPA") in nomes
        tem_pop = normalizar_path_text("População") in nomes

        if tem_casos and tem_mapa and tem_pop:
            return base

    raise FileNotFoundError(
        "Não encontrei a pasta-base contendo "
        "'Casos-TB-2001-2024', 'DADOS-BASE-MAPA' e 'População'."
    )


def encontrar_pasta(base, nome_esperado):
    base = Path(base)

    direto = base / nome_esperado
    if direto.exists():
        return direto

    alvo = normalizar_path_text(nome_esperado)

    for p in base.iterdir():
        if p.is_dir() and normalizar_path_text(p.name) == alvo:
            return p

    raise FileNotFoundError(f"Pasta não encontrada: {nome_esperado}")


def procurar_arquivo(pasta, padroes):
    pasta = Path(pasta)
    arquivos = [p for p in pasta.rglob("*") if p.is_file()]

    for arq in arquivos:
        nome = normalizar_path_text(arq.name)

        for padrao in padroes:
            padrao_norm = normalizar_path_text(padrao)
            if re.search(padrao_norm, nome):
                return arq

    print(f"\nArquivos encontrados em {pasta}:")
    for arq in arquivos:
        print(" -", arq.name)

    raise FileNotFoundError(
        f"Nenhum arquivo encontrado em {pasta} para os padrões: {padroes}"
    )


def salvar_fig(nome, pasta_saida):
    path = pasta_saida / nome
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    print("Salvo em:", path)


def periodo_quinquenal(ano):
    if 2001 <= ano <= 2005:
        return "2001–2005"
    if 2006 <= ano <= 2010:
        return "2006–2010"
    if 2011 <= ano <= 2015:
        return "2011–2015"
    if 2016 <= ano <= 2019:
        return "2016–2019"
    if 2020 <= ano <= 2024:
        return "2020–2024"
    return np.nan


# ============================================================
# 2. LOCALIZAÇÃO DOS ARQUIVOS
# ============================================================

BASE_DIR = encontrar_base_dir()

DIR_CASOS = encontrar_pasta(BASE_DIR, "Casos-TB-2001-2024")
DIR_GEO = encontrar_pasta(BASE_DIR, "DADOS-BASE-MAPA")
DIR_POP = encontrar_pasta(BASE_DIR, "População")

ARQ_CASOS_SP = DIR_CASOS / "Casos-TB-SP-NOVO.csv"
ARQ_CASOS_RJ = DIR_CASOS / "Casos-TB-RJ-NOVO.csv"

if not ARQ_CASOS_SP.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {ARQ_CASOS_SP}")

if not ARQ_CASOS_RJ.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {ARQ_CASOS_RJ}")

ARQ_POPULACAO = DIR_POP / "PopulaçãoCorrigida_interpolada_CAGR_FINAL.xlsx"

if not ARQ_POPULACAO.exists():
    ARQ_POPULACAO = procurar_arquivo(
        DIR_POP,
        [
            r"populacao.*cagr.*final.*\.xlsx$",
            r"populacao.*interpolada.*\.xlsx$",
            r"populacao.*corrigida.*\.xlsx$",
            r".*\.xlsx$"
        ]
    )

ARQ_GEO_RJ = DIR_GEO / "RJ_geojs-33-mun.json"
ARQ_GEO_SP = DIR_GEO / "SP_geojs-35-mun.json"

if not ARQ_GEO_RJ.exists():
    ARQ_GEO_RJ = procurar_arquivo(
        DIR_GEO,
        [
            r"rj.*geo.*\.json$",
            r"33.*mun.*\.json$",
            r"rj.*\.json$"
        ]
    )

if not ARQ_GEO_SP.exists():
    ARQ_GEO_SP = procurar_arquivo(
        DIR_GEO,
        [
            r"sp.*geo.*\.json$",
            r"35.*mun.*\.json$",
            r"sp.*\.json$"
        ]
    )

OUT_DIR = BASE_DIR / "saida_exploratoria_TB"
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_57 = OUT_DIR / "secao_5_7_analise_exploratoria_espacial"
OUT_57.mkdir(parents=True, exist_ok=True)

print("Pasta-base:", BASE_DIR)
print("Casos SP:", ARQ_CASOS_SP)
print("Casos RJ:", ARQ_CASOS_RJ)
print("População:", ARQ_POPULACAO)
print("Geo RJ:", ARQ_GEO_RJ)
print("Geo SP:", ARQ_GEO_SP)
print("Saída seção 5.7:", OUT_57)


# ============================================================
# 3. RECONSTRUÇÃO DO PAINEL MUNICÍPIO-ANO
# ============================================================

def read_csv_casos(path):
    last_error = None

    for enc in ["latin1", "cp1252", "utf-8-sig", "utf-8"]:
        try:
            df = pd.read_csv(
                path,
                sep=";",
                encoding=enc,
                dtype=str,
                engine="python",
                quoting=csv.QUOTE_NONE,
                keep_default_na=False,
                on_bad_lines="warn"
            )

            df.columns = [
                str(c).replace('"', "").replace("'", "").strip()
                for c in df.columns
            ]

            empty_cols = []
            for c in df.columns:
                vals = df[c].astype(str).str.strip()
                if vals.eq("").all():
                    empty_cols.append(c)

            if empty_cols:
                df = df.drop(columns=empty_cols)

            return df

        except Exception as e:
            last_error = e

    raise last_error


def clean_count_series(s):
    s = s.astype(str).str.strip()

    s = (
        s.str.replace('"', "", regex=False)
         .str.replace("'", "", regex=False)
         .str.strip()
    )

    s = s.replace({
        "-": "0",
        "": np.nan,
        "nan": np.nan,
        "NaN": np.nan,
        "None": np.nan
    })

    s = (
        s.str.replace(" ", "", regex=False)
         .str.replace(".", "", regex=False)
         .str.replace(",", ".", regex=False)
    )

    return pd.to_numeric(s, errors="coerce").fillna(0).astype(int)


def carregar_casos(path, estado):
    raw = read_csv_casos(path).copy()

    print(f"\nLendo casos {estado}:")
    print(path)
    print("Colunas encontradas:")
    print(raw.columns.tolist())

    col_mun = None

    for c in raw.columns:
        if "MUNICIPIO" in norm_text(c):
            col_mun = c
            break

    if col_mun is None:
        for c in raw.columns:
            serie = (
                raw[c]
                .astype(str)
                .str.replace('"', "", regex=False)
                .str.replace("'", "", regex=False)
                .str.strip()
            )

            n_match = serie.str.match(r"^\d{6}\s+.+", na=False).sum()

            if n_match >= 5:
                col_mun = c
                break

    if col_mun is None:
        display(raw.head(10))
        raise ValueError(f"Não encontrei coluna de município no arquivo: {path}")

    print("Coluna de município identificada:", col_mun)

    raw = raw[raw[col_mun].notna()].copy()

    raw[col_mun] = (
        raw[col_mun]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )

    raw = raw[~raw[col_mun].str.upper().isin(["TOTAL", ""])].copy()

    extracted = raw[col_mun].str.extract(r"^\s*(\d{6})\s+(.+?)\s*$")

    raw["cod_mun6"] = extracted[0]
    raw["municipio"] = extracted[1]

    raw = raw[raw["cod_mun6"].notna()].copy()

    raw["municipio"] = (
        raw["municipio"]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )

    year_cols = []

    for c in raw.columns:
        c_str = str(c).replace('"', "").replace("'", "").strip()

        if c_str.isdigit():
            ano = int(c_str)
            if ano in YEARS:
                year_cols.append(c)

    if len(year_cols) == 0:
        display(raw.head(10))
        raise ValueError(f"Não encontrei colunas de anos em: {path}")

    for c in year_cols:
        raw[c] = clean_count_series(raw[c])

    long = raw[["cod_mun6", "municipio"] + year_cols].melt(
        id_vars=["cod_mun6", "municipio"],
        var_name="ano",
        value_name="casos"
    )

    long["ano"] = (
        long["ano"]
        .astype(str)
        .str.replace('"', "", regex=False)
        .str.replace("'", "", regex=False)
        .astype(int)
    )

    long["estado"] = estado
    long["municipio"] = long["municipio"].map(norm_text)

    return long[["estado", "cod_mun6", "municipio", "ano", "casos"]]


def clean_pop_series(s):
    if pd.api.types.is_numeric_dtype(s):
        return pd.to_numeric(s, errors="coerce").round()

    txt = s.astype(str).str.strip()

    txt = (
        txt.str.replace('"', "", regex=False)
           .str.replace("'", "", regex=False)
           .str.replace(" ", "", regex=False)
    )

    mask_brasil = txt.str.contains(",", regex=False)

    txt_brasil = (
        txt[mask_brasil]
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False)
    )

    txt_outros = txt[~mask_brasil]

    out = pd.Series(index=s.index, dtype="float")
    out.loc[mask_brasil] = pd.to_numeric(txt_brasil, errors="coerce")
    out.loc[~mask_brasil] = pd.to_numeric(txt_outros, errors="coerce")

    return out.round()


def encontrar_sheet_estado(xlsx_path, estado):
    xl = pd.ExcelFile(xlsx_path)
    sheets = xl.sheet_names

    if estado == "RJ":
        termos = ["RJ", "RIO DE JANEIRO"]
    elif estado == "SP":
        termos = ["SP", "SAO PAULO", "SÃO PAULO"]
    else:
        termos = [estado]

    for sh in sheets:
        sh_norm = norm_text(sh)
        if any(norm_text(t) in sh_norm for t in termos):
            return sh

    print("Abas disponíveis:", sheets)
    raise ValueError(f"Não encontrei aba para {estado} em {xlsx_path}")


def identificar_coluna_codigo(df):
    for c in df.columns:
        cn = norm_text(c)
        if "COD" in cn:
            return c

    for c in df.columns:
        serie = df[c].astype(str)
        n_match = serie.str.extract(r"(\d{6})")[0].notna().sum()
        if n_match >= 5:
            return c

    return None


def identificar_coluna_municipio(df):
    for c in df.columns:
        cn = norm_text(c)
        if "MUNICIPIO" in cn:
            return c

    for c in df.columns:
        cn = norm_text(c)
        if cn in ["NOME", "NOME_MUNICIPIO", "NM_MUN", "NM_MUNICIPIO"]:
            return c

    return None


def carregar_populacao(path):
    partes = []

    for estado in ["RJ", "SP"]:
        sheet = encontrar_sheet_estado(path, estado)

        print(f"\nLendo população {estado}:")
        print("Aba:", sheet)

        df = pd.read_excel(path, sheet_name=sheet).copy()

        col_cod = identificar_coluna_codigo(df)
        col_mun = identificar_coluna_municipio(df)

        if col_cod is None:
            raise ValueError(f"Não encontrei coluna de código municipal na aba {sheet}")

        if col_mun is None:
            raise ValueError(f"Não encontrei coluna de município na aba {sheet}")

        df["cod_mun6"] = df[col_cod].astype(str).str.extract(r"(\d{6})")[0]
        df["municipio"] = df[col_mun].map(norm_text)

        year_cols = []

        for c in df.columns:
            c_str = str(c).strip()

            if c_str.isdigit():
                ano = int(c_str)
                if ano in YEARS:
                    year_cols.append(c)

        if len(year_cols) == 0:
            raise ValueError(f"Não encontrei colunas de anos na aba {sheet}")

        long = df[["cod_mun6", "municipio"] + year_cols].melt(
            id_vars=["cod_mun6", "municipio"],
            var_name="ano",
            value_name="populacao"
        )

        long["ano"] = long["ano"].astype(int)
        long["populacao"] = clean_pop_series(long["populacao"])
        long["estado"] = estado

        partes.append(
            long[["estado", "cod_mun6", "municipio", "ano", "populacao"]]
        )

    return pd.concat(partes, ignore_index=True)


casos_sp = carregar_casos(ARQ_CASOS_SP, "SP")
casos_rj = carregar_casos(ARQ_CASOS_RJ, "RJ")

casos = pd.concat([casos_sp, casos_rj], ignore_index=True)

populacao = carregar_populacao(ARQ_POPULACAO)

painel = casos.merge(
    populacao,
    on=["estado", "cod_mun6", "ano"],
    how="outer",
    suffixes=("_casos", "_pop"),
    indicator=True
)

painel["municipio"] = painel["municipio_casos"].combine_first(painel["municipio_pop"])
painel["casos"] = painel["casos"].fillna(0).astype(int)

painel = painel.sort_values(["estado", "cod_mun6", "ano"]).reset_index(drop=True)

painel["incidencia_100mil"] = np.where(
    painel["populacao"].gt(0),
    painel["casos"] / painel["populacao"] * 100000,
    np.nan
)

painel["log_pop"] = np.where(
    painel["populacao"].gt(0),
    np.log(painel["populacao"]),
    np.nan
)

painel["periodo"] = painel["ano"].apply(periodo_quinquenal)

painel = painel[
    [
        "estado",
        "cod_mun6",
        "municipio",
        "ano",
        "periodo",
        "casos",
        "populacao",
        "log_pop",
        "incidencia_100mil",
        "_merge"
    ]
]

painel = painel[painel["ano"].isin(YEARS)].copy()

painel["incidencia_ma3"] = (
    painel
    .sort_values(["estado", "cod_mun6", "ano"])
    .groupby(["estado", "cod_mun6"])["incidencia_100mil"]
    .transform(lambda x: x.rolling(window=3, min_periods=1, center=True).mean())
)

painel["log_casos_mais1"] = np.log1p(painel["casos"])
painel["log_inc_mais1"] = np.log1p(painel["incidencia_100mil"])

print("\nPainel reconstruído:")
display(
    painel.groupby("estado")
    .agg(
        linhas=("casos", "size"),
        municipios=("cod_mun6", "nunique"),
        ano_min=("ano", "min"),
        ano_max=("ano", "max"),
        casos_total=("casos", "sum"),
        incidencia_media=("incidencia_100mil", "mean")
    )
    .reset_index()
)

painel.to_csv(
    OUT_57 / "painel_reconstruido_secao_5_7.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 4. CARREGAMENTO DAS MALHAS MUNICIPAIS
# ============================================================

import geopandas as gpd

def carregar_malha(path, estado):
    gdf = gpd.read_file(path).copy()

    if "id" not in gdf.columns:
        raise ValueError(f"A malha {path} não tem coluna 'id'.")

    gdf["cod_mun6"] = gdf["id"].astype(str).str[:6]

    if "name" in gdf.columns:
        gdf["municipio_geo"] = gdf["name"].map(norm_text)
    else:
        gdf["municipio_geo"] = ""

    gdf["estado"] = estado

    if gdf.crs is None:
        gdf = gdf.set_crs("EPSG:4326")

    gdf = gdf[gdf.geometry.notna()].copy()

    return gdf


geo_rj = carregar_malha(ARQ_GEO_RJ, "RJ")
geo_sp = carregar_malha(ARQ_GEO_SP, "SP")

print("\nMalhas carregadas:")
display(
    pd.DataFrame({
        "estado": ["RJ", "SP"],
        "municipios_malha": [
            geo_rj["cod_mun6"].nunique(),
            geo_sp["cod_mun6"].nunique()
        ]
    })
)

geos = {
    "RJ": geo_rj,
    "SP": geo_sp
}


# ============================================================
# 5. BASES AGREGADAS PARA ANÁLISE ESPACIAL
# ============================================================

# ------------------------------------------------------------
# 5.1 Agregado municipal do período completo
# ------------------------------------------------------------

mun_periodo_total = (
    painel
    .groupby(["estado", "cod_mun6", "municipio"], as_index=False)
    .agg(
        casos_acumulados=("casos", "sum"),
        pop_pessoa_ano=("populacao", "sum"),
        pop_media=("populacao", "mean"),
        incidencia_media_anual=("incidencia_100mil", "mean"),
        incidencia_mediana_anual=("incidencia_100mil", "median"),
        incidencia_max_anual=("incidencia_100mil", "max"),
        anos_com_casos=("casos", lambda x: (x > 0).sum()),
        anos_zero=("casos", lambda x: (x == 0).sum())
    )
)

mun_periodo_total["incidencia_periodo_100mil_pessoa_ano"] = (
    mun_periodo_total["casos_acumulados"] /
    mun_periodo_total["pop_pessoa_ano"] * 100000
)

mun_periodo_total["log_casos_acumulados_mais1"] = np.log1p(
    mun_periodo_total["casos_acumulados"]
)

mun_periodo_total["log_inc_periodo_mais1"] = np.log1p(
    mun_periodo_total["incidencia_periodo_100mil_pessoa_ano"]
)

# ------------------------------------------------------------
# 5.2 Agregado por períodos
# ------------------------------------------------------------

mun_periodos = (
    painel
    .dropna(subset=["periodo"])
    .groupby(["estado", "cod_mun6", "municipio", "periodo"], as_index=False)
    .agg(
        casos_total=("casos", "sum"),
        pop_pessoa_ano=("populacao", "sum"),
        incidencia_media_anual=("incidencia_100mil", "mean"),
        incidencia_mediana_anual=("incidencia_100mil", "median")
    )
)

mun_periodos["incidencia_periodo_100mil_pessoa_ano"] = (
    mun_periodos["casos_total"] /
    mun_periodos["pop_pessoa_ano"] * 100000
)

mun_periodos["log_inc_periodo_mais1"] = np.log1p(
    mun_periodos["incidencia_periodo_100mil_pessoa_ano"]
)

mun_periodo_total.to_csv(
    OUT_57 / "base_municipal_periodo_total.csv",
    index=False,
    encoding="utf-8-sig"
)

mun_periodos.to_csv(
    OUT_57 / "base_municipal_periodos.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 6. MATRIZES DE VIZINHANÇA QUEEN
# ============================================================

from libpysal.weights import Queen

def construir_pesos_queen(estado):
    gdf = geos[estado].copy()

    gdf = gdf.sort_values("cod_mun6").reset_index(drop=True)

    ids = gdf["cod_mun6"].astype(str).tolist()

    w = Queen.from_dataframe(gdf, ids=ids)
    w.transform = "r"

    gdf["n_vizinhos"] = gdf["cod_mun6"].map(w.cardinalities)

    islands = list(w.islands)

    resumo = {
        "estado": estado,
        "municipios": len(gdf),
        "media_vizinhos": gdf["n_vizinhos"].mean(),
        "mediana_vizinhos": gdf["n_vizinhos"].median(),
        "min_vizinhos": gdf["n_vizinhos"].min(),
        "max_vizinhos": gdf["n_vizinhos"].max(),
        "municipios_sem_vizinhos": len(islands),
        "codigos_sem_vizinhos": "; ".join(islands) if len(islands) > 0 else ""
    }

    # edgelist
    linhas = []

    for origem, vizinhos in w.neighbors.items():
        for destino in vizinhos:
            linhas.append({
                "estado": estado,
                "origem": origem,
                "destino": destino
            })

    edgelist = pd.DataFrame(linhas)

    return gdf, w, pd.DataFrame([resumo]), edgelist


geo_rj_w, w_rj, resumo_w_rj, edges_rj = construir_pesos_queen("RJ")
geo_sp_w, w_sp, resumo_w_sp, edges_sp = construir_pesos_queen("SP")

resumo_vizinhanca = pd.concat([resumo_w_rj, resumo_w_sp], ignore_index=True)
arestas_vizinhanca = pd.concat([edges_rj, edges_sp], ignore_index=True)

print("\nResumo das matrizes de vizinhança queen:")
display(resumo_vizinhanca)

resumo_vizinhanca.to_csv(
    OUT_57 / "resumo_matriz_vizinhanca_queen.csv",
    index=False,
    encoding="utf-8-sig"
)

arestas_vizinhanca.to_csv(
    OUT_57 / "arestas_matriz_vizinhanca_queen.csv",
    index=False,
    encoding="utf-8-sig"
)

# Mapa diagnóstico do número de vizinhos
for estado, gdf_w in {"RJ": geo_rj_w, "SP": geo_sp_w}.items():
    fig, ax = plt.subplots(figsize=(9, 9))

    gdf_w.plot(
        column="n_vizinhos",
        ax=ax,
        legend=True,
        linewidth=0.2,
        edgecolor="black"
    )

    ax.set_title(f"Número de vizinhos queen por município — {estado}")
    ax.axis("off")

    salvar_fig(f"mapa_numero_vizinhos_queen_{estado}.png", OUT_57)


# ============================================================
# 7. FUNÇÕES ESPACIAIS: MORAN, LISA E GI*
# ============================================================

from esda.moran import Moran, Moran_Local
from esda.getisord import G_Local

def obter_geo_estado(estado):
    if estado == "RJ":
        return geo_rj.copy()
    if estado == "SP":
        return geo_sp.copy()
    raise ValueError("Estado inválido.")


def merge_geo_dados(estado, dados):
    gdf = obter_geo_estado(estado)

    dados_estado = dados[dados["estado"] == estado].copy()

    gdf = gdf.merge(
        dados_estado,
        on=["estado", "cod_mun6"],
        how="left"
    )

    return gdf


def construir_w_para_gdf(gdf, transform="r"):
    gdf = gdf.sort_values("cod_mun6").reset_index(drop=True)
    ids = gdf["cod_mun6"].astype(str).tolist()

    w = Queen.from_dataframe(gdf, ids=ids)
    w.transform = transform

    return w


def calcular_moran_global(gdf, variavel, estado, contexto, transform="r"):
    gdf = gdf.dropna(subset=[variavel]).copy()

    if len(gdf) < 5:
        return {
            "estado": estado,
            "contexto": contexto,
            "variavel": variavel,
            "n": len(gdf),
            "I_Moran": np.nan,
            "p_sim": np.nan,
            "z_sim": np.nan,
            "erro": "menos de 5 observações"
        }

    y = gdf[variavel].astype(float).values

    if np.nanstd(y) == 0:
        return {
            "estado": estado,
            "contexto": contexto,
            "variavel": variavel,
            "n": len(gdf),
            "I_Moran": np.nan,
            "p_sim": np.nan,
            "z_sim": np.nan,
            "erro": "variável constante"
        }

    w = construir_w_para_gdf(gdf, transform=transform)

    moran = Moran(y, w, permutations=PERMUTATIONS)

    return {
        "estado": estado,
        "contexto": contexto,
        "variavel": variavel,
        "n": len(gdf),
        "I_Moran": moran.I,
        "p_sim": moran.p_sim,
        "z_sim": moran.z_sim,
        "erro": ""
    }


def classificar_lisa(q, p, alpha=ALPHA):
    if pd.isna(p) or p > alpha:
        return "Não significativo"

    if q == 1:
        return "Alto-Alto"
    if q == 2:
        return "Baixo-Alto"
    if q == 3:
        return "Baixo-Baixo"
    if q == 4:
        return "Alto-Baixo"

    return "Não classificado"


def calcular_lisa(gdf, variavel, estado, contexto):
    gdf = gdf.dropna(subset=[variavel]).copy()
    gdf = gdf.sort_values("cod_mun6").reset_index(drop=True)

    y = gdf[variavel].astype(float).values

    if len(gdf) < 5 or np.nanstd(y) == 0:
        gdf["lisa_I"] = np.nan
        gdf["lisa_p"] = np.nan
        gdf["lisa_q"] = np.nan
        gdf["classe_lisa"] = "Não calculado"
        gdf["estado_analise"] = estado
        gdf["contexto_analise"] = contexto
        gdf["variavel_analise"] = variavel
        return gdf

    w = construir_w_para_gdf(gdf, transform="r")

    lisa = Moran_Local(y, w, permutations=PERMUTATIONS)

    gdf["lisa_I"] = lisa.Is
    gdf["lisa_p"] = lisa.p_sim
    gdf["lisa_q"] = lisa.q

    gdf["classe_lisa"] = [
        classificar_lisa(q, p, ALPHA)
        for q, p in zip(gdf["lisa_q"], gdf["lisa_p"])
    ]

    gdf["estado_analise"] = estado
    gdf["contexto_analise"] = contexto
    gdf["variavel_analise"] = variavel

    return gdf


def classificar_gi(z, p, alpha=ALPHA):
    if pd.isna(p) or p > alpha:
        return "Não significativo"

    if z > 0:
        return "Hot spot"

    if z < 0:
        return "Cold spot"

    return "Não significativo"


def calcular_getis_ord_gi(gdf, variavel, estado, contexto):
    gdf = gdf.dropna(subset=[variavel]).copy()
    gdf = gdf.sort_values("cod_mun6").reset_index(drop=True)

    y = gdf[variavel].astype(float).values

    if len(gdf) < 5 or np.nanstd(y) == 0:
        gdf["gi_z"] = np.nan
        gdf["gi_p"] = np.nan
        gdf["classe_gi"] = "Não calculado"
        gdf["estado_analise"] = estado
        gdf["contexto_analise"] = contexto
        gdf["variavel_analise"] = variavel
        return gdf

    # Para Gi*, é comum usar pesos binários.
    w = construir_w_para_gdf(gdf, transform="B")

    gi = G_Local(
        y,
        w,
        star=True,
        permutations=PERMUTATIONS
    )

    z = getattr(gi, "Zs", None)

    if z is None:
        z = getattr(gi, "z_sim", np.full(len(gdf), np.nan))

    gdf["gi_z"] = z
    gdf["gi_p"] = gi.p_sim

    gdf["classe_gi"] = [
        classificar_gi(zv, pv, ALPHA)
        for zv, pv in zip(gdf["gi_z"], gdf["gi_p"])
    ]

    gdf["estado_analise"] = estado
    gdf["contexto_analise"] = contexto
    gdf["variavel_analise"] = variavel

    return gdf


def salvar_tabela_sem_geometria(gdf, path):
    cols = [c for c in gdf.columns if c != "geometry"]
    gdf[cols].to_csv(path, index=False, encoding="utf-8-sig")


def plot_mapa_categorico(gdf, coluna, titulo, nome):
    fig, ax = plt.subplots(figsize=(9, 9))

    gdf.plot(
        column=coluna,
        ax=ax,
        categorical=True,
        legend=True,
        linewidth=0.2,
        edgecolor="black",
        missing_kwds={
            "color": "lightgrey",
            "label": "Sem dado"
        }
    )

    ax.set_title(titulo, fontsize=13)
    ax.axis("off")

    salvar_fig(nome, OUT_57)


def plot_mapa_numerico(gdf, coluna, titulo, nome):
    fig, ax = plt.subplots(figsize=(9, 9))

    gdf.plot(
        column=coluna,
        ax=ax,
        legend=True,
        linewidth=0.2,
        edgecolor="black",
        missing_kwds={
            "color": "lightgrey",
            "label": "Sem dado"
        }
    )

    ax.set_title(titulo, fontsize=13)
    ax.axis("off")

    salvar_fig(nome, OUT_57)


# ============================================================
# 8. MORAN GLOBAL
# ============================================================

resultados_moran = []

# ------------------------------------------------------------
# 8.1 Moran para período completo
# ------------------------------------------------------------

variaveis_periodo_total = [
    "incidencia_periodo_100mil_pessoa_ano",
    "log_inc_periodo_mais1",
    "log_casos_acumulados_mais1"
]

for estado in ["RJ", "SP"]:
    gdf = merge_geo_dados(estado, mun_periodo_total)

    for var in variaveis_periodo_total:
        resultados_moran.append(
            calcular_moran_global(
                gdf=gdf,
                variavel=var,
                estado=estado,
                contexto="2001–2024",
                transform="r"
            )
        )

# ------------------------------------------------------------
# 8.2 Moran por período
# ------------------------------------------------------------

for estado in ["RJ", "SP"]:
    for periodo in sorted(mun_periodos["periodo"].dropna().unique()):
        dados = mun_periodos[mun_periodos["periodo"] == periodo].copy()
        gdf = merge_geo_dados(estado, dados)

        for var in ["incidencia_periodo_100mil_pessoa_ano", "log_inc_periodo_mais1"]:
            resultados_moran.append(
                calcular_moran_global(
                    gdf=gdf,
                    variavel=var,
                    estado=estado,
                    contexto=periodo,
                    transform="r"
                )
            )

# ------------------------------------------------------------
# 8.3 Moran anual: incidência bruta e incidência média móvel 3 anos
# ------------------------------------------------------------

for estado in ["RJ", "SP"]:
    for ano in YEARS:
        dados = painel[painel["ano"] == ano].copy()
        gdf = merge_geo_dados(estado, dados)

        for var in ["incidencia_100mil", "incidencia_ma3", "log_inc_mais1"]:
            resultados_moran.append(
                calcular_moran_global(
                    gdf=gdf,
                    variavel=var,
                    estado=estado,
                    contexto=str(ano),
                    transform="r"
                )
            )

moran_global_57 = pd.DataFrame(resultados_moran)

print("\nMoran Global — resultados:")
display(moran_global_57.head(20))

moran_global_57.to_csv(
    OUT_57 / "moran_global_resultados.csv",
    index=False,
    encoding="utf-8-sig"
)

# Gráficos de Moran Global anual
for var in ["incidencia_100mil", "incidencia_ma3", "log_inc_mais1"]:
    dplot = moran_global_57[
        (moran_global_57["variavel"] == var) &
        (moran_global_57["contexto"].isin([str(y) for y in YEARS]))
    ].copy()

    dplot["ano"] = dplot["contexto"].astype(int)

    plt.figure(figsize=(10, 5))

    for estado, sub in dplot.groupby("estado"):
        sub = sub.sort_values("ano")
        plt.plot(
            sub["ano"],
            sub["I_Moran"],
            marker="o",
            linewidth=2,
            label=estado
        )

    plt.axhline(0, linestyle="--")
    plt.title(f"Moran Global anual — {var}")
    plt.xlabel("Ano")
    plt.ylabel("I de Moran")
    plt.grid(alpha=0.3)
    plt.legend()

    salvar_fig(f"serie_moran_global_anual_{var}.png", OUT_57)


# ============================================================
# 9. LISA — PERÍODO COMPLETO E ANOS SELECIONADOS
# ============================================================

lisa_tabelas = []

# ------------------------------------------------------------
# 9.1 LISA para período completo
# ------------------------------------------------------------

for estado in ["RJ", "SP"]:
    gdf_base = merge_geo_dados(estado, mun_periodo_total)

    for var in ["incidencia_periodo_100mil_pessoa_ano", "log_inc_periodo_mais1"]:
        contexto = "2001–2024"

        lisa_gdf = calcular_lisa(
            gdf=gdf_base,
            variavel=var,
            estado=estado,
            contexto=contexto
        )

        lisa_tabelas.append(lisa_gdf.drop(columns="geometry").copy())

        nome_base = f"lisa_{estado}_{var}_2001_2024"

        salvar_tabela_sem_geometria(
            lisa_gdf,
            OUT_57 / f"{nome_base}.csv"
        )

        try:
            lisa_gdf.to_file(
                OUT_57 / f"{nome_base}.geojson",
                driver="GeoJSON"
            )
        except Exception:
            pass

        plot_mapa_categorico(
            lisa_gdf,
            "classe_lisa",
            f"LISA — {var}\n{estado}, {contexto}",
            f"mapa_{nome_base}.png"
        )

# ------------------------------------------------------------
# 9.2 LISA para anos selecionados
# ------------------------------------------------------------

for estado in ["RJ", "SP"]:
    for ano in ANOS_SELECIONADOS:
        dados = painel[painel["ano"] == ano].copy()
        gdf_base = merge_geo_dados(estado, dados)

        for var in ["incidencia_100mil", "incidencia_ma3"]:
            contexto = str(ano)

            lisa_gdf = calcular_lisa(
                gdf=gdf_base,
                variavel=var,
                estado=estado,
                contexto=contexto
            )

            lisa_tabelas.append(lisa_gdf.drop(columns="geometry").copy())

            nome_base = f"lisa_{estado}_{var}_{ano}"

            salvar_tabela_sem_geometria(
                lisa_gdf,
                OUT_57 / f"{nome_base}.csv"
            )

            try:
                lisa_gdf.to_file(
                    OUT_57 / f"{nome_base}.geojson",
                    driver="GeoJSON"
                )
            except Exception:
                pass

            plot_mapa_categorico(
                lisa_gdf,
                "classe_lisa",
                f"LISA — {var}\n{estado}, {ano}",
                f"mapa_{nome_base}.png"
            )

lisa_resultados_57 = pd.concat(lisa_tabelas, ignore_index=True)

lisa_resultados_57.to_csv(
    OUT_57 / "lisa_resultados_consolidados.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nResumo das classes LISA:")
display(
    lisa_resultados_57
    .groupby(["estado_analise", "contexto_analise", "variavel_analise", "classe_lisa"])
    .size()
    .reset_index(name="n")
)


# ============================================================
# 10. GETIS-ORD GI*
# ============================================================

gi_tabelas = []

# ------------------------------------------------------------
# 10.1 Gi* para período completo
# ------------------------------------------------------------

for estado in ["RJ", "SP"]:
    gdf_base = merge_geo_dados(estado, mun_periodo_total)

    for var in ["incidencia_periodo_100mil_pessoa_ano", "log_inc_periodo_mais1"]:
        contexto = "2001–2024"

        gi_gdf = calcular_getis_ord_gi(
            gdf=gdf_base,
            variavel=var,
            estado=estado,
            contexto=contexto
        )

        gi_tabelas.append(gi_gdf.drop(columns="geometry").copy())

        nome_base = f"getis_ord_gi_{estado}_{var}_2001_2024"

        salvar_tabela_sem_geometria(
            gi_gdf,
            OUT_57 / f"{nome_base}.csv"
        )

        try:
            gi_gdf.to_file(
                OUT_57 / f"{nome_base}.geojson",
                driver="GeoJSON"
            )
        except Exception:
            pass

        plot_mapa_categorico(
            gi_gdf,
            "classe_gi",
            f"Getis-Ord Gi* — {var}\n{estado}, {contexto}",
            f"mapa_{nome_base}.png"
        )

# ------------------------------------------------------------
# 10.2 Gi* para anos selecionados
# ------------------------------------------------------------

for estado in ["RJ", "SP"]:
    for ano in ANOS_SELECIONADOS:
        dados = painel[painel["ano"] == ano].copy()
        gdf_base = merge_geo_dados(estado, dados)

        var = "incidencia_ma3"
        contexto = str(ano)

        gi_gdf = calcular_getis_ord_gi(
            gdf=gdf_base,
            variavel=var,
            estado=estado,
            contexto=contexto
        )

        gi_tabelas.append(gi_gdf.drop(columns="geometry").copy())

        nome_base = f"getis_ord_gi_{estado}_{var}_{ano}"

        salvar_tabela_sem_geometria(
            gi_gdf,
            OUT_57 / f"{nome_base}.csv"
        )

        try:
            gi_gdf.to_file(
                OUT_57 / f"{nome_base}.geojson",
                driver="GeoJSON"
            )
        except Exception:
            pass

        plot_mapa_categorico(
            gi_gdf,
            "classe_gi",
            f"Getis-Ord Gi* — {var}\n{estado}, {ano}",
            f"mapa_{nome_base}.png"
        )

gi_resultados_57 = pd.concat(gi_tabelas, ignore_index=True)

gi_resultados_57.to_csv(
    OUT_57 / "getis_ord_gi_resultados_consolidados.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nResumo das classes Getis-Ord Gi*:")
display(
    gi_resultados_57
    .groupby(["estado_analise", "contexto_analise", "variavel_analise", "classe_gi"])
    .size()
    .reset_index(name="n")
)


# ============================================================
# 11. MAPAS NUMÉRICOS DAS VARIÁVEIS USADAS
# ============================================================

for estado in ["RJ", "SP"]:
    gdf = merge_geo_dados(estado, mun_periodo_total)

    plot_mapa_numerico(
        gdf,
        "incidencia_periodo_100mil_pessoa_ano",
        f"Incidência do período por 100 mil pessoa-ano\n{estado}, 2001–2024",
        f"mapa_incidencia_periodo_100mil_{estado}_2001_2024.png"
    )

    plot_mapa_numerico(
        gdf,
        "log_inc_periodo_mais1",
        f"log(1 + incidência do período)\n{estado}, 2001–2024",
        f"mapa_log_incidencia_periodo_{estado}_2001_2024.png"
    )

    plot_mapa_numerico(
        gdf,
        "log_casos_acumulados_mais1",
        f"log(1 + casos acumulados)\n{estado}, 2001–2024",
        f"mapa_log_casos_acumulados_{estado}_2001_2024.png"
    )


# ============================================================
# 12. EXPORTAÇÃO FINAL EM EXCEL
# ============================================================

excel_57 = OUT_57 / "secao_5_7_analise_exploratoria_espacial.xlsx"

resumo_lisa = (
    lisa_resultados_57
    .groupby(["estado_analise", "contexto_analise", "variavel_analise", "classe_lisa"])
    .size()
    .reset_index(name="n")
)

resumo_gi = (
    gi_resultados_57
    .groupby(["estado_analise", "contexto_analise", "variavel_analise", "classe_gi"])
    .size()
    .reset_index(name="n")
)

with pd.ExcelWriter(excel_57) as writer:
    resumo_vizinhanca.to_excel(writer, sheet_name="vizinhanca_queen", index=False)
    moran_global_57.to_excel(writer, sheet_name="moran_global", index=False)
    resumo_lisa.to_excel(writer, sheet_name="resumo_lisa", index=False)
    resumo_gi.to_excel(writer, sheet_name="resumo_gi", index=False)
    mun_periodo_total.to_excel(writer, sheet_name="base_periodo_total", index=False)
    mun_periodos.to_excel(writer, sheet_name="base_periodos", index=False)

print("\n============================================================")
print("SEÇÃO 5.7 FINALIZADA")
print("Análise exploratória espacial concluída.")
print("Resultados salvos em:")
print(OUT_57)
print("\nArquivo Excel principal:")
print(excel_57)
print("============================================================")
```
